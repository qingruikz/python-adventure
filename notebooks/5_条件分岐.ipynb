## **5.1 プログラムの流れ：文と制御構造**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/5-2.png)


**制御構造の種類**

1.  **逐次 (Sequential):** 文を**上から下に書かれた順番通りに**実行していく、最も基本的な流れだ。これまで我々が主に使ってきたもの。（例：変数に値を代入し、次にそれを表示する）
2.  **選択 (Selection):** ある条件が満たされているかどうかに応じて、実行する文を選ぶ。（例：「もし雨なら傘を持つ」）
3.  **反復 (Iteration / Repetition):** ある条件が満たされている間、同じ文のブロックを何度も繰り返し実行する。（例：「目標歩数に達するまで歩く」）

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/5-3.png)


![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/5-4.png)


**「選択」と「条件分岐」の関係**

- **選択 (Selection):** これは、プログラムの流れをコントロールする**考え方や概念**の一つだ。「複数の道がある中で、状況（条件）に応じて進む道を選ぶ」という、プログラムの振る舞いの種類を表す、少し抽象的な言葉なんだ。目的地に行くのに、「もし晴れてたら山道、雨なら舗装路」みたいに行き方を選ぶ、その**考え方自体**が「選択」にあたる。
- **条件分岐 (Conditional Branching):** これは、その「選択」という考え方を、**Pythonというプログラミング言語で具体的に実現するための書き方（構文や仕組み）** のことなんだ。さっきの例で言えば、「晴れ」「雨」という条件を見て、「山道へ行け」「舗装路へ行け」と書かれた **具体的な道路標識や指示**が「条件分岐」にあたる。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/5-5.png)


**「反復 」と「繰り返し」の関係**

- **反復 (Iteration / Repetition):** これもプログラムの流れをコントロールする**考え方や概念**の一つ。「同じような作業を何度も行う」という、プログラムの振る舞いの種類を表す抽象的な言葉だ。例えば、ラジオ体操で「腕を回す運動を8回繰り返す」という**考え方自体**が「反復」にあたる。
- **繰り返し (Loop):** これは、「反復」という考え方を、**Pythonで具体的に実現するための書き方（構文や仕組み）**のことだ。ラジオ体操で言えば、「イチ、ニ、サン、シ…」と**具体的な号令や指示**が「繰り返しにあたる。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/5-6.png)


## **5.2 条件分岐の基本構造：`if`文とインデントの掟**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/5-7.png)


### **5.2.1 `if`文の基本形**

```python
if 条件式:
    条件式が真(True)のときに実行したい処理1
    条件式が真(True)のときに実行したい処理2
    # ... さらに処理が続く場合もある
```

**`if`文の基本的な書き方はこうだ**。
*   `if` というキーワードで始める。
*   続けて、結果が `True` (真) または `False` (偽) になる**条件式**を書く。
*   条件式の後には必ず**コロン `:`** を付ける。
*   次の行から、条件が `True` の場合に実行したい処理を**字下げ（インデント）**して書く。

### **5.2.2 ブロックとインデント**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/5-8.png)


![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/5-9.png)


**インデントのルール:**

*   同じブロック内の文は、**必ず同じ幅でインデント**する。
*   一般的に、**半角スペース4つ**でインデントするのが推奨されている（Pythonの公式スタイルガイドPEP 8）。
*   インデントが正しくないと、`IndentationError` というエラーが発生する。

実際に簡単な例を見てみよう。

````python
# コード例5-1
score = 80

if score >= 60:
    print("合格です！") # インデントされているので、score >= 60 が True の場合に実行される
    print("よく頑張りましたね！") # これも同じブロック


# インデントされていないので、if文の結果に関わらず常に実行される
print("処理を終了します。") 
````

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/5-10.png)


````python
# コード例5-2
score = 50

if score >= 60:
    print("合格です！")  # インデントされているので、score >= 60 が True の場合に実行される
    print("よく頑張りましたね！")  # これも同じブロック


# インデントされていないので、if文の結果に関わらず常に実行される
print("処理を終了します。")
````

## **5.3 条件式：True/False を判定する式**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/5-16.png)


### **5.3.1 比較演算子**

二つの値を比較し、その結果を `True` か `False` で返すのが**比較演算子**だ。

| 演算子 | 意味                      | 例 (`a=5`, `b=10`) | 結果   |
| :----- | :------------------------ | :----------------- | :----- |
| `==`   | 等しい                    | `a == 5`           | `True` |
| `!=`   | 等しくない                | `a != b`           | `True` |
| `>`    | より大きい                | `b > a`            | `True` |
| `<`    | より小さい                | `a < b`            | `True` |
| `>=`   | 以上 (より大きいか等しい) | `a >= 5`           | `True` |
| `<=`   | 以下 (より小さいか等しい) | `b <= 10`          | `True` |

**最重要注意点：** 「等しい」を比較するのはイコール2つの `==` だ！ イコール1つの `=` は**代入**を意味するので、`if` の条件式で間違って使うと、意図しない動作やエラーの原因になるぞ！

````python
# コード例5-3: 等しいか (==)
age = 20

if age == 20:
    print("年齢は20歳です。")
````

````python
# コード例5-4: 等しくないか (!=)
if age != 18:
    print("年齢は18歳ではありません。")
````

````python
# コード例5-5: 以上か (>=)
if age >= 18:
    print("成人です。")
````

````python
# コード例5-6: わざとエラーを起こしてみる（間違いやすい例）：= を使うとエラーまたは意図しない動作
if age = 20: # これは代入なのでエラー！ (SyntaxError)
    print("エラーになるはず")
````

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/5-17.png)


### **5.3.2 `in` 演算子 (復習)**

リストや文字列の中に、特定の要素や部分文字列が含まれているかを調べる `in` や `not in` も、結果が `True`/`False` になるので条件式として使える。

````python
# コード例5-7: リストに含まれるか (in)
fruits = ["リンゴ", "バナナ", "オレンジ"]

if "バナナ" in fruits:
    print("リストにバナナは含まれています。")
````

````python
# コード例5-8: 文字列に含まれないか (not in)
message = "Hello Python"
if "Java" not in message:
    print("メッセージにJavaは含まれていません。")
````

### **5.3.3 真偽値 (`True`, `False`)**

比較演算子や `in` 演算子の結果として返ってくる `True` や `False`。これらの値そのものを**真偽値 (Boolean value)** と呼び、データ型は `bool` だったね。`if`文は、条件式が最終的に `True` になるかどうかを見ているんだ。

````python
# コード例5-9
is_ok = True
is_error = False

if is_ok:  # is_ok は True なので実行される
    print("状態はOKです。")
````

````python
# コード例5-10
if is_error:  # is_error は False なので、何も表示されない
    print("エラー発生")
````

### **5.3.4 論理演算子 (`and`, `or`, `not`)**

複数の条件を組み合わせたい場合がある。「年齢が18歳以上 **かつ** 運転免許を持っている」とか、「科目が数学 **または** 理科である」とかね。そんな時に使うのが**論理演算子**だ。

*   `A and B`: A **かつ** B (A と B の両方が `True` の場合に `True`)
*   `A or B`: A **または** B (A と B の少なくとも一方が `True` の場合に `True`)
*   `not A`: A **でない** (A が `True` なら `False`、`False` なら `True` を返す)

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/5-18.png)


````python
# コード例5-11: and（両方の条件を満たすか）
age = 25
has_license = True

if age >= 18 and has_license == True:
    print("運転可能です。")
````

````python
# コード例5-12: or（どちらか一方の条件を満たすか）
score_math = 80
score_english = 55
if score_math >= 60 or score_english >= 60:
    print("少なくともどちらかの科目は合格点です。")
````

````python
# コード例5-13: not（条件を否定するか）
is_raining = False
if not is_raining:
    print("雨は降っていません。傘は不要です。")
````

## **5.4 分岐構文のバリエーション：もっと柔軟な条件分岐**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/5-20.png)


### **5.4.1 `if-else` 構文：条件が偽の場合の処理**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/5-21.png)


````python
# コード例5-14
score = 50

if score >= 60:
    print("合格です！")  # score < 60 なので実行されない
else:
    print("残念、不合格です。")  # score >= 60 が False なのでこちらが実行される
    print("次回頑張りましょう。")  # elseブロックの処理

print("--- 判定終了 ---")
````

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/5-22.png)


### **5.4.2 `if` のみの構文 (復習)**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/5-23.png)


````python
# コード例5-15
temperature = 32

if temperature >= 30:
    print("猛暑日です！水分補給を忘れずに。")  # 実行される

# else がないので、temperature < 30 の場合は何も表示されない
print("今日の天気予報でした。")  # これは if とは関係なく実行される
````

### **5.4.3 `if-elif-else` 構文：複数の条件で分岐**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/5-24.png)


`elif` は "else if" の略で、「そうでなくて、もし〜なら」という意味を持つ。`elif` はいくつでも繋げることができる。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/5-25.png)


````python
# コード例5-16
score = 75

if score >= 80:
    print("評価は 優 です。")
elif score >= 60:  # 80点未満 かつ 60点以上 の場合
    print("評価は 良 です。")  # score = 75 はこの条件に合致
elif score >= 40:  # 60点未満 かつ 40点以上 の場合
    print("評価は 可 です。")
else:  # 40点未満の場合
    print("評価は 不可 です。")

print("--- 成績評価終了 ---")
````

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/5-26.png)


### **5.4.4 `if` 文のネスト：入れ子になった条件分岐**

`if` 文のブロックの中に、さらに `if` 文を入れることもできる。これを **ネスト (Nest, 入れ子)** と呼ぶ。これにより、より複雑な条件判定が可能になる。

````python
# コード例5-17
age = 15
is_student = True

if age < 18:
    print("未成年です。")
    # さらに学生かどうかで分岐 (ネストされたif)
    if is_student:
        print("学割が適用されます。")
    else:
        print("学割は適用されません。")
else:
    print("成人です。")
````

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/5-27.png)


### **5.4.5 三項演算子：簡単な `if-else` を1行で**

最後に、非常に簡単な `if-else` を1行で簡潔に書くための特別な書き方を紹介しよう。**三項演算子 (Ternary Operator)** と呼ばれるものだ。

`条件が真の場合の値 if 条件式 else 条件が偽の場合の値`

という形式で書く。

````python
# コード例5-18
age = 20

# 通常の if-else
if age >= 18:
    status = "成人"
else:
    status = "未成年"

print(f"あなたは {status} です。")
````

````python
# コード例5-19: 三項演算子を使った場合
status = "成人" if age >= 18 else "未成年"

print(f"あなたは {status} です。")
````

## **5.5 今回の冒険のまとめ**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/5-29.png)


この章では、次のことを学びました。

*   **プログラムの流れ:**
    *   プログラムは**文 (`Statement`)** の集まり。
    *   **制御構造 (`Control Structure`)** が文の実行順序を決める（逐次、選択、反復）。
*   **条件分岐の基本:**
    *   `if 条件式:` で条件分岐を開始。コロン `:` を忘れずに。
    *   **インデント**で処理の**ブロック**を定義。半角スペース4つが推奨。タブとの混在は避ける (`IndentationError`, `TabError`)。
*   **条件式:**
    *   `if` の後に書く、結果が `True`/`False` になる式。
    *   **比較演算子:** `==` (等しい、`=`は代入！), `!=`, `>`, `<`, `>=`, `<=`。
    *   **`in` 演算子:** 要素の存在確認。
    *   **真偽値 (`bool`)**: `True` と `False`。
    *   **論理演算子:** `and` (かつ), `or` (または), `not` (でない)。
*   **分岐構文のバリエーション:**
    *   `if-else`: 条件が偽の場合の処理。
    *   `if` のみ: 条件が真の場合のみ処理。
    *   `if-elif-else`: 複数の条件で分岐 (`elif` はいくつでも可)。上から順に評価される。
    *   ネスト (`if` の中に `if`): 複雑な条件。深くしすぎない注意が必要。
    *   三項演算子: `真の場合の値 if 条件式 else 偽の場合の値` (簡単な `if-else` を1行で)。
