**7. 絶対パスと相対パス：住所の書き方２種類**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/11-11.png)

*   **絶対パス (Absolute Path):** ルートディレクトリ（整理棚の一番上 `/` や `C:\`）から**目的地までの全ての道順を省略せずに書く**方法。どこから見ても必ず同じ場所を指す、いわば「完全な住所」。
    *   例 (Mac/Linux/Colab): `/Users/Kato/Documents/Data/sales.csv`
    *   例 (Windows): `C:\Users\Kato\Documents\Data\sales.csv`
*   **相対パス (Relative Path):** **今いる場所（カレントワーキングディレクトリ）を基準にして**、目的地までの道順を書く方法。いわば「ここから見て、あの角を曲がって…」というような案内方法だね。
    *   例：もし今 `ドキュメント` フォルダにいるなら、さっきの `レポート.docx` への相対パスは `./大学/経済学/レポート.docx` （今いる場所 `ドキュメント` 内の `大学` フォルダ、その中の `経済学` サブフォルダにあるファイル `レポート.docx` を指す）となる。
    *   また、`../` (ドットドットスラッシュ) は「**一つ上の階層（親ディレクトリ）に戻る**」という意味で、これも相対パスでよく使われる。例えば、`大学` から一つ上の `ドキュメント` フォルダにある `memo.txt` を指す場合、絶対パスなら `/ユーザー/ドキュメント/memo.txt`、相対パスなら `../memo.txt` となる。

### **11.2.1 CSVファイルの読み込み (`pd.read_csv`)**

**CSVファイルの中身のイメージ (例: `data.csv`)**

**テキストエディタで見た場合:**

```
名前,年齢,好きな果物
佐藤,20,リンゴ
高橋,20,バナナ
加藤,38,オレンジ
```

**表計算ソフトや `pandas` で読み込んだ場合のイメージ:**

| 名前 | 年齢 | 好きな果物 |
| :--- | :--: | :--------- |
| 佐藤 |  20  | リンゴ     |

| 加藤 |  38  | オレンジ   |

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/11-25.png)

````python
# コード例11-5: 読み込むCSVファイルのパスを確認
file1_path = CWD + "/データ/少子化率と高齢化率.csv"
print(f"読み込むCSVファイル: {file1_path}")
````

````python
# コード例11-6: CSVファイルを読み込む
df1 = pd.read_csv(
    file1_path,  # ファイルのパス（住所）を指定
    encoding="shift-jis",  # 日本語を含むファイルは文字コード指定が重要！ (Shift_JISやUTF-8など)
    skiprows=8,  # ファイルの先頭8行を読み飛ばす（ヘッダーなどが不要な場合）
    na_values=["-", "***"],  # これらを欠損値(NaN)として扱うように指定
    thousands=",",  # 数字の桁区切り文字(,)を無視して数値として読み込む
)
````

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/11-26.png)

- **最初の引数:** これは必須。読み込むファイルのパス（住所）を指定する。。`pd.read_csv()` の最初の引数はファイルパス、というのは決まっている。
- **`encoding="shift-jis"`:** ここからはキーワード引数だ。順番は関係ない。`encoding` という名前の引数（オプション）に `"shift-jis"` という値を指定している。これで「文字コードはShift_JISでお願いします」と明確に指示できる。日本語が含まれるCSVでは**非常に重要**だ！ ファイルがどの文字コード（Shift_JIS, UTF-8, CP932など）で保存されているか正しく指定しないと、**文字化け**してしまう。どの文字コードか分からない場合は、いくつか試してみる必要があるかもしれない。
- **`skiprows=8`:** `skiprows` というキーワード引数に `8` を指定。「先頭8行をスキップしてね」という意味だ。
- **`na_values=[...]`:** `na_values` というキーワード引数に、欠損値として扱いたい文字のリスト `['-', '***']` を指定。データがないことを示す表現（例えば `-` や `***`、空白セルなど）はファイルによって様々だ。これを指定しておくと、pandasがそれらを統一的に**欠損値 `NaN`** として扱ってくれるので、後の処理が楽になる。
- **`thousands=','`:** `thousands` というキーワード引数に `,` を指定。「桁区切りのカンマは無視して数値にしてね」という意味。`1,234` のようなカンマ区切りの数字を、ちゃんと数値 `1234` として読み込むための指定。

````python
# コード例11-18

# --- ファイル1: 2000年までの所得データ ---
# csvファイルをデータフレームとして読み込む
df2 = pd.read_csv(
    CWD + "/データ/課税対象所得_2000年まで.csv",  # csvファイルのパス
    encoding="shift-jis",  # 日本語対応
    skiprows=8,  # 先頭の何行分をスキップか
    na_values=["-"],  # 欠損値として扱う値を追加
    thousands=",",  # 1000ごとの桁区切りの記号を指定する引数
)

# いらない列を削除
df2.drop(columns=["/項目", "注釈"], inplace=True)

# 列名を変更
df2.rename(
    columns={
        "C120110_課税対象所得【千円】": "課税対象所得",
        "地域 コード": "地域コード",
        "調査年 コード": "調査年コード",
    },
    inplace=True,
)

# 「調査年コード」を通常の年度に変更
df2["調査年コード"] = df2["調査年コード"] // 1000000

# 先頭5行を表示
df2.head()
````

### **11.4.2 横方向の結合 (`pd.merge`)**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/11-54.png)


**キー (Key) の重要性**

`pd.merge()` を使う上で最も重要なのが、二つの表を結びつけるための**共通の鍵（キー）となる列**だ。今回の例で言えば、「どの地域の」「どの年の」データなのかを示す**「地域コード」と「調査年コード」**が、二つの表を結びつける共通のキーになる。`pd.merge()` は、このキーが一致する行同士を見つけて、それらの情報を横に結合してくれるんだ。

**基本的な使い方: `pd.merge(左の表, 右の表, on=['キー列1',...])`**

*   `pd.merge()` 関数の最初の2つの引数に、結合したい `DataFrame`（左側に来る表、右側に来る表）を指定する。
*   `on` 引数に、結合のキーとなる**共通の列名**を指定する。キーが複数の列にまたがる場合は、`on=['キー列1', 'キー列2']` のようにリストで指定する。

````python
# コード例11-22: データの結合
df5 = pd.merge(df1, df4, on=["地域コード", "調査年"])

# 先頭5行を表示
df5.head()
````

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/11-55.png)


![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/11-56.png)


**重複列の対処**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/11-57.png)


````python
# コード例11-23: データの結合（事前に片方から共通の列を削除）
df4.drop(columns=["地域", "調査年"], inplace=True)
df6 = pd.merge(df1, df4, on=["地域コード", "調査年コード"])

# 先頭5行を表示
df6.head()
````

**結合方法の指定 (`how`)**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/11-58.png)


*   `how='inner'` (`how` 引数を指定しない場合のデフォルト設定): 両方の表に**共通して存在するキー**の行だけを残す。（内部結合）
*   `how='left'`: **左の表（最初の引数）のキーをすべて**残す。右の表に対応するキーがない場合は、右の表由来の列は欠損値 `NaN` になる。（左外部結合）
*   `how='right'`: **右の表（2番目の引数）のキーをすべて**残す。左の表に対応するキーがない場合は、左の表由来の列は欠損値 `NaN` になる。（右外部結合）
*   `how='outer'`: **両方の表のキーをすべて**残す。片方にしかキーがない場合は、対応する表由来の列は欠損値 `NaN` になる。（完全外部結合）

## **11.5 綺麗になったデータの保存**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/11-60.png)


**`.to_csv()` メソッドの使い方**

`DataFrame変数.to_csv('保存したいファイルパス', encoding='文字コード', index=False)`

*   `DataFrame変数`: 保存したいDataFrameが入っている変数名。
*   `.to_csv()`: DataFrameをCSV形式で保存するためのメソッド。
*   **最初の引数**: 保存先の**ファイルパス**（フォルダの場所とファイル名）を文字列で指定する。これが**必須**だ。
*   `encoding` 引数: 保存するファイルの**文字コード**を指定する。日本語を含むデータをExcelなど他のソフトでも正しく開けるようにするためには、`'utf_8_sig'` や `'shift_jis'` (または `'cp932'`) を指定するのが一般的だ。`'utf_8_sig'` は比較的新しいExcelでも文字化けしにくいのでおすすめだね。
*   `index` 引数: DataFrameの**インデックス**（一番左の行番号）をファイルに**書き出すかどうか**を指定する。通常、このインデックスは分析には不要なことが多いので、`index=False` と指定して**書き出さない**ようにするのが一般的だ。これを指定しないと、余計な列がCSVファイルに追加されてしまうことがある

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/11-60-a.png)

````python
# コード例11-24-a
# 1. 保存先のフォルダパスを定義
output_folder = CWD + "/出力ファイル"

# 2. フォルダが存在しない場合にのみ作成する
os.makedirs(output_folder, exist_ok=True)
````

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/11-60-b.png)

````python
# コード例11-24-b
# 3. 保存するファイルのフルパスを定義
output_filepath = output_folder + "/処理済みデータ.csv"

# 4. 実際にファイルを保存
df6.to_csv(output_filepath, encoding="shift_jis", index=False)
````

**`DataFrame変数.to_excel` の基本的な使い方:** DataFrameをExcelファイルに書き出すメソッド。

`DataFrame変数.to_excel('保存したいファイルパス', sheet_name='シート名', index=False)`
