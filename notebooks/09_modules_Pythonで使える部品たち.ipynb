## **9.1 コードの保管庫：実験ノート(`.ipynb`)と呪文書(`.py`)**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/9-2.png)

**`.ipynb` ファイルと `.py` ファイルの違い**

*   **`.ipynb` (魔法の実験ノート):**
    *   **特徴:** コード、実行結果、説明文を一緒に保存できる。セル単位で実行可能。
    *   **用途:** データ分析、試行錯誤、学習、プレゼンテーションなど、**対話的な作業**に向いている。我々がこれまでやってきたのは、まさにこれだ。
*   **`.py` (純粋な呪文書):**
    *   **特徴:** Pythonのコード**だけ**が書かれた、純粋なテキストファイル。実行結果や説明文は含まれない。
    *   **用途:** Webアプリケーションや自動化ツールのように、プログラムとして完成させて実行する場合や、そして今回学ぶ**再利用可能な部品（モジュール）を作る**場合に使う。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/9-3.png)

## **9.2 モジュール、パッケージ、ライブラリの関係：工房と道具の比喩**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/9-4.png)

1.  **モジュール (Module)**
    *   プラス、マイナス、星形など、様々な種類のドライバーをまとめて一つのケースに入れた **「ドライバーセット」** のようなものだ。Pythonの世界では、関連する関数や変数をまとめた**1つの `.py` ファイル**がこれにあたる。
2.  **パッケージ (Package)**
    *   複数の関連する道具セット（モジュール）を、整理して入れておくための**一つの工具箱**。この工具箱が**フォルダ**にあたる。Pythonの世界では、関連する複数のモジュール（`.py` ファイル）を**1つのフォルダにまとめたもの**がこれにあたる。
3.  **ライブラリ (Library) = 工房 (機能の集合)**
    *   たくさんの工具箱（パッケージ）や、特別な専門道具セット（モジュール）が揃った、**巨大な工房全体**。工房には、最初から備え付けの基本的な工具箱（**標準ライブラリ**）と、後から自分で注文して取り寄せた最新鋭の専門工具箱（**外部ライブラリ**）がある。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/9-5.png)

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/9-6.png)

**関係性のまとめ**

*   **階層:** **モジュール (ドライバーセット) < パッケージ (工具箱) < ライブラリ (工房)**
*   **構成:**
    *   **モジュール:** 1つの `.py` ファイル。中に関数や変数がたくさん。
    *   **パッケージ:** モジュールを含むフォルダ。
    *   **ライブラリ:** パッケージやモジュールを集めたもの。

## **9.4 パッケージ：「工具箱」でドライバーセットを整理**

### **9.4.1 パッケージとは？**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/9-22.png)

*   **目的:** 関連するモジュール群をグループ化し、**階層的に管理**する。これにより、大規模なプロジェクトでもコードの構造が分かりやすくなり、モジュール名の衝突も避けやすくなる。
*   **仕組み:** Pythonがディレクトリをパッケージとして認識するためには、そのディレクトリ内に `__init__.py` という名前のファイル（中身は空でも良い）を置くのが伝統的な方法だ（最近のPythonでは必須ではない場合もあるが、互換性や慣習で置くことが多い）。このファイルは、パッケージがインポートされたときに最初に実行される。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/9-23.png)

**今回作成する自作パッケージ `my_toolbox` のファイルツリー構造**

```
my_toolbox/                  <-- パッケージのルートフォルダ（工具箱本体）
├── __init__.py                <-- このフォルダがパッケージである印
├── string_tools.py            <-- 文字列操作用のモジュール（ドライバーセットA）
└── math_tools/                  <-- 数学ツール用のサブフォルダ（小箱）
    ├── __init__.py              <-- このフォルダがサブパッケージである印
    └── advanced_calc.py         <-- 高度な計算用モジュール（専門ドライバーセットB）
```

**ステップ1：フォルダ（パッケージとサブパッケージ）の作成**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/9-26.png)

**`os.makedirs()` 関数の使い方と便利なオプション**

*   `os.makedirs('作成したいフォルダのパス')`: 指定したパスのフォルダを作成する。この関数の賢いところは、もし途中のフォルダ（この例では `my_toolbox`）がまだ存在していなくても、それらも**全部まとめて**作ってくれる点だ。
*   `exist_ok=True`: これは「もし指定したフォルダが**既に存在していても (exist is ok)、エラーにしないでね**」という、非常に便利な「おまじない」のような引数だ。もしこれが無いと、同じコードを2回実行しただけで、「フォルダが既にあります！」というエラーが出てプログラムが止まってしまう。これを付けておけば、何度実行しても安全にフォルダの存在を保証できるんだ。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/9-27.png)

````python
# コード例9-11
import os

# パッケージのフォルダ構造を作成
os.makedirs("my_toolbox/math_tools", exist_ok=True)
print("パッケージ用のフォルダ構造を作成しました。")
````

*   `datetime.datetime`: **日付と時刻の両方**を扱うための道具。
*   `datetime.date`: **日付だけ**を扱うための道具。

````python
# コード例9-23: 標準ライブラリ datetime の使用例
import datetime

# 今日の日付を取得
today = datetime.date.today()
print(f"今日の日付: {today}")

# 現在の日時を取得
now = datetime.datetime.now()
print(f"現在の日時: {now}")
print(f"現在の時: {now.hour}時")  # 時だけ取り出すこともできる
````
