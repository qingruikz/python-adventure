**関数（処理の部品化）のメリットまとめ**

1.  **楽ちん（再利用性）:** 同じ処理を何度も書く必要がなくなり、関数の名前を呼び出すだけで使える。
2.  **分かりやすい（可読性）:** プログラム全体の見通しが良くなり、どこで何が行われているかが理解しやすくなる。
3.  **直しやすい（保守性）:** 処理内容を変更したい場合、関数の定義部分だけを修正すれば、その関数を呼び出している全ての箇所に反映される。修正漏れやミスを防げる。
4.  **安全（独立性）:** 関数内の処理は、基本的には他の部分から独立しているため、影響範囲を限定できる（後述）。

### **8.1.3 関数の定義**

Pythonで関数の設計図（定義）を書くには、`def` を使う。これが一番シンプルな形だ。

```python
def 関数名():
    # この関数が呼び出されたときに実行される処理
    処理1
    処理2
    # ... (インデントされている間が関数の範囲)
```

*   `def` と書く。
*   自分で考えた**関数名**を書く（変数名と同じルール。処理内容が分かる名前が良い）。
*   丸括弧 `()` を書く（今は空っぽでOK）。
*   **コロン `:`** を書く。
*   次の行から、**インデント（スペース4つ推奨）**して、関数にさせたい処理を書く（これが関数の**本体**、**ブロック**）。

さっきの自己紹介関数の定義をもう一度見てみよう。

````python
# コード例8-3: 自己紹介を表示する関数 show_intro を定義する
def show_intro():
    print("--- 自己紹介 ---")
    print("こんにちは！佐藤拓海です。")
    print("ラーメンとガジェットが好きです！")
    print("----------------")


# このコードを実行しても、まだ何も表示されない。
# あくまで「こういう関数を作りましたよ」とPythonに教えただけ。
````

**`関数名()`**

### **8.2.1 引数 (Argument / Parameter)：関数への指示や材料**

**引数 (ひきすう)** とは、関数を呼び出すときに、**関数に渡す情報（データ）** のことだ。関数に「これを使って仕事してね！」と材料を渡すようなイメージだね。

関数を定義するときに、`()` の中に変数名を書くことで、「この関数はこういう情報を受け取りますよ」と宣言できる。この定義時に書かれる変数を**仮引数 (Parameter)** と呼ぶ。

そして、関数を呼び出すときに、`()` の中に具体的な値を書くことで、その値が関数内の仮引数に渡される。呼び出し時に渡す具体的な値を**実引数 (Argument)** と呼ぶ。（※仮引数と実引数を厳密に区別せず、両方「引数」と呼ぶことも多い）

````python
# コード例8-8
# name という仮引数を受け取る関数 greet を定義
def greet(name):  # この name が仮引数
    print(f"こんにちは、{name}さん！")  # 受け取った name を使う


# 関数を呼び出すときに実引数を渡す
greet("佐藤")  # 実引数 "佐藤" が仮引数 name に渡される
greet("高橋")  # 実引数 "高橋" が仮引数 name に渡される
````

### **8.2.3 戻り値 (`return`)：関数から結果を受け取る**

関数に処理をさせた**結果**を、呼び出した場所で受け取りたい時に使うのが `return` 文だ。関数の中で `return 返したい値` と書くと、その時点で関数の実行は終わり、`返したい値` が関数の「成果物」として呼び出し元に返される。

````python
# コード例8-12
# 2つの数字を受け取り、足し算の結果を返す関数 add
def add(a, b):
    result = a + b
    return result  # 計算結果を返す！


# 関数を呼び出して、戻り値を変数 sum_result に入れる
sum_result = add(5, 3)  # add(5, 3) が実行され、戻り値 8 が sum_result に入る
print(f"5 + 3 の結果は: {sum_result}")

# 戻り値を直接 print 文の中で使うこともできる
print(f"3 + 7 の結果は: {add(3, 7)}")
````

### **8.2.4 関数呼び出しの `()` の意味**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/8-23.png)


**ステップ1： `=` の右側が先に評価される**

- まず、Pythonが `sum_result = add(5, 3)` という文に出会ったとき、基本ルールとして **`=` の右側にある式、つまり `add(5, 3)` の部分を先に評価（計算して値にする）** しようとするんだ。

**ステップ2： `関数名()` は「実行して結果（戻り値）を出す」魔法**

- ここが重要だ！`=` の右側にある `add(5, 3)` は、ただの「`add`」という名前ではない。後ろに付いている `()` は、「**`add` という名前の関数を、引数 `5` と `3` で実行せよ！そして、その実行結果（戻り値）をここに持ってきなさい！**」という強力な魔法の指示なんだ。これは**関数呼び出し演算子 (Call Operator)** と呼ばれる。まるで関数の「実行ボタン」を押すようなものだね。

**ステップ3： 関数が実行され、戻り値 `8` が現れる**

- この `()` の魔法に従って、Python は `add` 関数の中の処理を実行する。

  1. 引数 `5` と `3` が関数の中に入る。
  2. 中で `5 + 3` が計算されて `8` になる。
  3. `return 8` によって、値 8 が関数の**戻り値**としてポンっと出てくる。

**ステップ4： 出てきた「結果」が変数に代入される**

- さあ、これで `=` の右側の `add(5, 3)` を評価した結果が、具体的な**値 `8`** になったわけだ。
- そして最後に、この**計算結果の値 `8`** が、`=` の左側にある変数 `sum_result` に**代入**される。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/8-24.png)


- **`sum_result = add(5, 3)`:** `()` があるので `add` 関数を**実行**し、その**戻り値 `8`** を `sum_result` に代入する。
- **`my_func = add`:** `()` がないので `add` 関数を**実行せず**、関数**そのもの（関数オブジェクト）**に `my_func` という別の名前を付ける。

````python
# コード例8-13: 関数オブジェクトの代入例
def add(a, b):
    return a + b


# () を付けずに代入 -> 関数オブジェクトが代入される
function_alias = add
print(f"function_alias は: {function_alias}")
````

````python
# コード例8-14
type(function_alias)
````

````python
# コード例8-15: 別名をつけた関数を呼び出すには、やはり () が必要
result = function_alias(10, 20)
print(f"別名で呼び出した結果: {result}")
````

### **8.2.5 戻り値がない？ (`None`)**

もし関数の中に `return` 文がなかったり、`return` の後に何も書かなかったりすると、その関数は**自動的に `None` という特別な値を返す**。`None` は「何もないよ」という意味だ。

````python
# コード例8-16
def just_print(message):
    print(message)
    # return がない


result = just_print("メッセージ表示")  # 関数は実行されるが…
print(f"just_print の戻り値は: {result}")
````

### **8.2.6 `return` で関数を途中で抜ける**

`return` 文が実行されると、その時点で関数の処理は**即座に終了**し、値が返る。`if` 文と組み合わせれば、特定の条件で関数の処理を途中で打ち切ることができる。

````python
# コード例8-17
def check_positive(number):
    if number <= 0:
        print("0以下の数値です。")
        return  # 条件に合わないので、ここで関数を終了 (Noneが返る)

    # この下の処理は number > 0 の場合だけ実行される
    print(f"{number} は正の数です！")
    return number  # 正の数であることを示すために、その数を返す


# "5 は正の数です！" と表示され、5が返る
check_positive(5)
````

````python
# コード例8-18: "0以下の数値です。" と表示され、Noneが返る
check_positive(-2)
````

### **8.2.7 関数の連携：関数から別の関数を呼ぶ**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/8-27.png)


````python
# コード例8-19
# 足し算関数 (再掲)
def add(a, b):
    return a + b


# 平均値関数 (中で add を使う)
def average(x, y):
    total = add(x, y)  # ★ add関数を呼び出す！
    return total / 2


# 平均値を計算して表示
avg_result = average(8, 12)
print(f"8と12の平均は: {avg_result}")  # 10.0
````

### **8.3.1 デフォルト引数：引数を省略可能にする**

関数を定義するとき、引数に「もし呼び出し側で指定されなかったら、この値を使ってね」という**デフォルト値（初期値）**を設定できる。`仮引数名=デフォルト値` のように書く。

````python
# コード例8-20
# greeting のデフォルト値を "Hi" に設定
def greet_with_default(name, greeting="Hi"):
    print(f"{greeting}, {name}!")


# greeting を省略 -> デフォルト値 "Hi" が使われる
greet_with_default("高橋さん")
````

````python
# コード例8-21: greeting を指定 -> 指定した値 "おはよう" が使われる
greet_with_default("高橋さん", "おはよう")
````

**注意:** デフォルト値を設定する引数は、設定しない引数の**後ろ**に書く必要がある。

````python
# コード例8-26: キーワード専用引数を持つ関数の定義
def create_character(name, *, job, level):
    # * の後ろにある job と level はキーワードでしか指定できない
    print(f"名前: {name}")
    print(f"職業: {job}")
    print(f"レベル: {level}")

# 正しい呼び出し方 (job と level はキーワードで渡すしかない)
create_character("佐藤", job="戦士", level=5)
````

### **8.3.3 可変長引数 (`*args`)：いくつでも引数を受け取る**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/8-33.png)


````python
# コード例8-27: 任意の個数の数値を受け取り、合計を返す関数
def sum_all(*args):  # 複数の引数がタプル args にまとめられる
    print(f"受け取った引数たち (argsというタプル): {args}")
    total = 0

    # args はタプルなので、forループで要素を一つずつ取り出せる
    for num in args:
        total += num
    return total
````

````python
# コード例8-28: 引数を2つ渡す
result1 = sum_all(1, 2)
print(f"合計1: {result1}")
````

````python
# コード例8-29: 引数を5つ渡す
result2 = sum_all(10, 20, 30, 40, 50)
print(f"合計2: {result2}")
````

````python
# コード例8-30: 引数を渡さない
result3 = sum_all()
print(f"合計3: {result3}")
````

### **8.3.4 辞書型の可変長引数 ( `**kwargs` )：キーワード引数をまとめて受け取る**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/8-35.png)


````python
# コード例8-31: 任意のキーワード引数を受け取り、表示する関数
def print_info(**kwargs):  # キーワード引数が辞書 kwargs にまとめられる
    print(f"受け取ったキーワード引数 (辞書): {kwargs}")
    for key, value in kwargs.items():
        print(f"キー'{key}' の値は '{value}' です。")


# キーワード引数を指定して呼び出す
print_info(name="高橋", age=20, city="東京")
````

### **8.3.5 引数の順番**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/8-37.png)


**関数を定義する際の、引数の正しい順番:**

1.  **通常の引数 (位置でもキーワードでも渡せる):** `def func(pos1, pos2):`
    - **渡し方:** 呼び出し側は、**位置で渡す**ことも、**引数の名前（キーワード）で渡す**ことも可能。
2.  **デフォルト引数:** `def func(pos1, default1='default'):`
    - **渡し方:** 通常の引数と同様、**位置でもキーワードでも**値を渡せる。**省略すると**デフォルト値が使われる。
3.  **可変長位置引数 (`*args`):** `def func(pos1, *args):`
    - **渡し方:** 通常の引数やデフォルト引数に割り当てられなかった、残りの**位置引数**が全て、タプルとしてこの引数にまとめられる。
4.  **キーワード専用引数 (デフォルト値なし/あり):** `def func(pos1, *, kw_only1, kw_only2='default'):`
    - **渡し方:** この引数は、**引数の名前（キーワード）でしか**値を渡すことができない。位置で渡そうとするとエラーになる。
5.  **可変長キーワード引数 (`**kwargs`):** `def func(pos1, **kwargs):`
    - **渡し方:** 呼び出しの際に**キーワードで渡す形式（引数名=値）**の引数のうち、他のどの仮引数にも対応しなかったものが、全て辞書としてこの引数にまとめられます。

### **8.3.8 高階関数 (`map`, `filter`)：関数を操る関数**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/8-42.png)


高階関数の代表例：

*   **`map(関数, イテラブル)`**:
    *   リストなどの**各要素**に対して、**引数で渡された関数**を適用した結果を生成。
    *   例：リストの全ての数値を文字列に変換したい → `map(str, [1, 2, 3])`
*   **`filter(関数, イテラブル)`**:
    *   リストなどの**各要素**に対して、**引数で渡された関数**（`True`/`False`を返すもの）を適用して **`True` になる要素だけ**を抽出。
    *   例：リストの中から負の数だけを取り出したい → `filter(lambda x: x < 0, [-1, 0, 1])`

これらの関数は、結果が特殊なデータで返ってくるので、通常は `list()` 関数を使ってリストに変換してから中身を確認することが多い。

````python
# コード例8-39
nums = [1, 2, 3, 4]
# mapで各要素を2乗
squares = list(map(lambda x: x**2, nums))
print(f"mapの結果: {squares}")  # [1, 4, 9, 16]
````

````python
# コード例8-40: filterで偶数のみ抽出
evens = list(filter(lambda x: x % 2 == 0, nums))
print(f"filterの結果: {evens}")  # [2, 4]
````

## **8.4 今回の冒険のまとめ**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/8-44.png)


この章では、次のことを学びました。

*   **関数の基礎:**

    *   関数のメリット: 再利用性、可読性、保守性向上（**部品化**）。
    *   使い方: `def 関数名(): ...` で**定義**し、`関数名()` で**呼び出す**。
    *   **ローカル変数**: 関数内で定義された変数は基本的にその関数内でのみ有効（スコープ）。
*   **引数と戻り値:**

    *   **引数**: 関数に情報を渡す仕組み（定義時: 仮引数、呼び出し時: 実引数）。
    *   **戻り値**: `return 値` で関数から結果を返す。なければ `None` が返る。`return` で関数は終了。
    *   関数の連携: 関数内から他の関数を呼び出せる。
*   **関数の応用：もっと柔軟な引数と戻り値:**

    *   **デフォルト引数:** `def func(arg="デフォルト値"):` のように定義時に初期値を設定でき、呼び出し時にその引数を**省略可能**にする
    *   **渡し方のスタイル:**
        *   **位置で渡す:** `func(値1, 値2)` のように、**順番**で値を渡す基本の方法。
        *   **名前（キーワード）で渡す:** `func(arg=値)` のように、**仮引数名**を指定して値を渡す。順番を自由に変えられ、コードが読みやすくなる。
    *   **キーワード専用引数:** `def func(*, arg):` のように定義すると、その引数は**名前（キーワード）でしか**値を渡せなくなる。
    *   **可変長引数:**
        *   **`*args` (可変長位置引数):** 任意の個数の**位置で渡された引数**を、一つの**タプル**としてまとめて受け取る。
        *   **`**kwargs` (可変長キーワード引数):** 任意の個数の**名前で渡された引数**を、一つの**辞書**としてまとめて受け取る。
    *   **複数の戻り値:** `return a, b` と書くと、値が**タプル `(a, b)` にまとめられて**返る。アンパック代入で別々の変数に受け取れる。
    *   **(発展) ラムダ式と高階関数:** `lambda` で一行の簡単な関数を作ったり、`map()` や `filter()` のように関数を引数として渡したりできる。
