### **4.2.1 要素の変更**

リストの特定の要素を別の値で上書きするのは簡単だ。インデックスを指定して、新しい値を代入するだけ。

```python
# コード例4-20

colors = ["赤", "青", "黄"]
print(f"変更前: {colors}")

# 要素の上書き：インデックス1を書き換え
colors[1] = "緑"
print(f"変更後: {colors}")
```

### **4.2.2 要素の追加**

新しい要素をリストに加えたい場合、いくつかのメソッドがある。
- `.append(要素)`: リストの**末尾**に要素を**1つ**追加。一番シンプル。
- `.insert(インデックス, 要素)`: 指定した**位置**に要素を挿入。
- `.extend(別のリスト)`: リストの末尾に**別のリストの全要素**を追加（合体）。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/4-18.png)


```python
# コード例4-21
my_list = [10, 20]
print(f"初期リスト: {my_list}")
```

```python
# コード例4-22: 末尾に追加
my_list.append(30)
print(my_list)
```

```python
# コード例4-23: 指定位置に挿入
my_list.insert(1, 15)  # インデックス1に挿入
print(my_list)
```

```python
# コード例4-24: 別のリストを結合
another_list = [40, 50]
my_list.extend(another_list)
print(my_list)
```

### **4.2.3 要素の削除**

要素を削除する方法も複数ある。状況に応じて使い分けよう。

- `.remove(値)`: 指定した**値**を持つ要素を**最初の1つだけ**削除。
- `.pop(インデックス)`: 指定した**位置**の要素を削除し、**その要素を返す**。インデックス省略時は末尾。
- `del リスト[インデックス]`: 指定した**位置**の要素を削除（値を返さない）。
- `del リスト[開始:終了]`: スライスで指定した範囲の要素を削除。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/4-20.png)


```python
# コード例4-25
numbers = [10, 20, 30, 20, 40]
print(numbers)
```

```python
# コード例4-26: 値で削除。最初の20が消える
numbers.remove(20)
print(numbers)
```

```python
# コード例4-27: 位置で削除して値を取得。インデックス2 (30) を削除
removed_value = numbers.pop(2)
print(f".pop(2)後のnumbers: {numbers}")
print(f"popで取り出した値: {removed_value}")
```

```python
# コード例4-28: delで位置指定削除。ンデックス0 (10) を削除
del numbers[0]
print(f"del後のnumbers: {numbers}")
```

```python
# コード例4-29: delとスライスで範囲削除。インデックス1以降を削除
del numbers[1:] 
print(f"del[1:]後のnumbers: {numbers}")
```

### **4.3.5 破壊的操作と非破壊的操作**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/4-27.png)


1. **破壊的操作 (Destructive Operation / In-place Operation):**
   - 実行すると、**操作対象の元のデータ（この場合はリスト）そのものを直接変更してしまう**操作。
   - In-place（インプレース）とは「その場で」という意味で、データが格納されている場所で直接変更が行われるイメージだ。
   - 例：`my_list.append(10)` を実行すると、`my_list` というリスト**自体**の末尾に `10` が追加される。
   - **これまで学んだ中での例:** 4.2節の操作全般 (`リスト[i]=値`, `.append()`, `.insert()`, `.extend()`, `.remove()`, `.pop()`, `del`, `.reverse()`) や、後で学ぶ `.sort()` メソッドなど。
2. **非破壊的操作 (Non-destructive Operation / Operation that creates a new object):**
   - 実行しても、**元のデータは一切変更せず**、操作の結果として**新しいデータ（新しいリスト、数値、文字列、True/Falseなど）を生成して返す**操作。
   - Non-destructive とは「破壊しない」という意味。元のデータは安全に保たれる。
   - 例：`new_list = list_a + list_b` を実行すると、`list_a` と `list_b` は変わらず、連結された**新しいリスト** `new_list` が作られる。
   - **これまで学んだ中での例:** 4.1節の操作 (`len()`, `[]`, `[:]`) や 4.3節の操作 (`max()`, `min()`, `sum()`, `+`, `*`, `in`, `not in`, `.split()`, `.join()`)、後で学ぶ `sorted()` 関数など。

### **4.3.6 リストのソート** 

リストの要素を並び替える方法。破壊的か非破壊的かで使い分ける。

- `.sort()` **メソッド (破壊的)**: リスト**自体**を並び替える。
- `sorted()` **関数 (非破壊的)**: 並び替えた**新しいリスト**を返す。

```python
# コード例4-44

# .sort() メソッド (破壊的): 元のリストが変わる
numbers1 = [30, 10, 50, 20]
print(f"ソート前 numbers1: {numbers1}")

# 昇順ソート
numbers1.sort()
print(f".sort() 後の numbers1: {numbers1}")

# 降順ソート
numbers1.sort(reverse=True)
print(f".sort(reverse=True) 後の numbers1: {numbers1}")
```

```python
# コード例4-45

# sorted() 関数 (非破壊的)
numbers2 = [30, 10, 50, 20]
print(f"ソート前 numbers2: {numbers2}")

# 新しい昇順リストを返す
sorted_asc = sorted(numbers2)
print(f"sorted() 結果: {sorted_asc}")
# 元のリストは変わらない
print(f"sorted() 後の numbers2: {numbers2}")

# 新しい降順リストを返す
sorted_desc = sorted(numbers2, reverse=True)
print(f"sorted(reverse=True) 結果: {sorted_desc}")
# 元のリストは変わらない
print(f"sorted() 後の numbers2: {numbers2}")
```

### **4.4.4 タプルの利点と使いどころ：なぜ「変更できない」が嬉しいのか？**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/4-40.png)


**タプルの利点**

1. **データの安全性・不変性の保証:**
   プログラムを書いていると、「このデータの組は絶対に変わってほしくない」という場面がある。例えば、設定値の組、座標、色のRGB値などだ。タプルを使えば、プログラムのどこかで誤ってその中身が書き換えられてしまう心配がない。**意図しない変更を防ぎ、データの完全性を保つ**ことができるんだ。これは、特に大規模なプログラムや複数人で開発する場合に重要になる。
2. **パフォーマンス（わずかな差だが）:**
   内部的な仕組みにより、一般的にリストよりもタプルの方が若干メモリ効率が良く、要素へのアクセス速度も速いとされることがある。ただし、これは通常、体感できるほどの大きな差ではなく、決定的な選択理由にはなりにくい。

  - 複数のデータを**順番に**格納でき、**変更可能 (mutable)** なデータ型。
  - 角括弧 `[]` で作成する (`[要素1, 要素2, ...]`)。
  - `len(リスト)` で要素数を取得。
  - インデックス `リスト[インデックス]` で要素にアクセス (`0`始まり、負のインデックス可)。
  - スライス `リスト[開始:終了]` で部分リストを取得 (終了は含まない)。
  - リストの中にリストを入れる**多重リスト** (`リスト[行][列]`) も可能。
- **リストの要素操作 (主に破壊的):**
  - 要素の変更: `リスト[インデックス] = 新しい値`。
  - 要素の追加:
    - `.append(要素)`: 末尾に1つ追加。
    - `.insert(インデックス, 要素)`: 指定位置に挿入。
    - `.extend(別のリスト)`: 末尾に別のリストの全要素を追加。
  - 要素の削除:
    - `.remove(値)`: 最初に見つかった指定の値を持つ要素を削除。
    - `.pop(インデックス)`: 指定位置の要素を削除し、その値を返す (省略時は末尾)。
    - `del リスト[インデックス]` または `del リスト[開始:終了]`: 指定位置または範囲の要素を削除。
  - 並び順の反転: `.reverse()`。
- **リストの応用機能と注意点:**
  - 数値リスト用関数: `max()`, `min()`, `sum()` (これらは非破壊的)。
  - 演算子: `+` (連結), `*` (繰り返し) は**新しいリスト**を生成する (**非破壊的**)。
  - 検索: `in`, `not in` で要素の存在を確認 (**非破壊的**)。
  - 文字列連携:
    - `文字列.split(区切り文字)`: 文字列を分割して**新しいリスト**を生成。
    - `区切り文字.join(文字列リスト)`: リスト要素を結合して**新しい文字列**を生成。
  - **破壊的 vs 非破壊的**: 元のリストを直接変更する操作 (`.append()`, `.sort()` など) と、変更せずに新しい値を返す操作 (`+`, `sorted()` など) の違いを理解することが重要。
  - ソート:
    - `.sort()`: リスト**自体**を並び替える (**破壊的**)。
    - `sorted(リスト)`: 並び替えた**新しいリスト**を返す (**非破壊的**)。
    - `reverse=True` オプションで降順ソート。
- **タプル (tuple)**:
  - 複数のデータを**順番に**格納できるが、**変更不可能 (immutable)**。
  - 丸括弧 `()` またはカンマ区切りで作成 (`(要素1, 要素2, ...)`)。
  - **注意:** 要素が1つのタプルは `(値,)` のように末尾にカンマが必要。
  - 変更を伴わない操作 (`len()`, インデックス `[]`, スライス `[:]`, `+`, `*`, `in`, `not in`) はリストと同様に可能。
  - 要素の変更・追加・削除は**エラー**になる。
  - 利点: データの安全性確保、など。「変更されたくないデータ」に適している。
