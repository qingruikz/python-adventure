## **16.2 学習スタイルで見る機械学習の主な種類**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/16-8.png)

**回帰分析の基本モデル (再掲):** $Y_i = \alpha + \beta_1 X_{i,1} + \beta_2 X_{i,2} + \dots + \beta_K X_{i,K} + \varepsilon_i$

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/16-9.png)

*   **ラベル (Label) / ターゲット (Target) / 目的変数:** 機械学習モデルが**予測しようとする対象**のことだ。回帰分析で言うところの**被説明変数 $Y_i$** が、これに相当する。
*   **特徴量 (Feature) / 説明変数 / 入力データ (Input Data):** 上記のラベルを予測するために**手がかりとなるデータ**のこと。回帰分析で言うところの**説明変数 $X_{i,1}, X_{i,2}, \dots, X_{i,K}$** がこれに相当する。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/16-10.png)

1.  **教師あり学習 (Supervised Learning):**
    *   コンピューターに「特徴量（問題）」と「それに対応する正解ラベル（答え）」が**セットになったデータ**を大量に与えて、その関係性を学習させる方法。まさに、解答付きのドリルで学ぶイメージだね。我々が経験した回帰分析は、この教師あり学習の代表例の一つだ。
2.  **教師なし学習 (Unsupervised Learning):**
    *   コンピューターに「正解ラベル」が**ない**データ（特徴量のみのデータ）を大量に与え、データの中に潜む構造やパターン、あるいはデータのグループ分けなどをコンピューター自身に見つけ出させる方法。お手本なしで、データそのものから学ぶ。
3.  **強化学習 (Reinforcement Learning):**
    *   コンピューター（エージェント）が、ある環境の中で様々な「行動」を試し、その結果として得られる「報酬」や「罰」というフィードバックをもとに、最適な行動戦略を学習していく方法。試行錯誤を通じて学ぶ。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/16-11.png)

### **16.2.1 教師あり学習 (Supervised Learning)：お手本（正解ラベル）から学ぶ**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/16-12.png)

*   **A) 回帰 (Regression):** 予測したい**正解ラベル (Label)** が、**連続的な数値**である場合。これは、まさに第14章で学んだ**回帰分析**と同じ考え方だ！ 例えば…
    *   明日の株価はいくらになるか？ (ラベル: 株価という数値)
    *   この家の適切な販売価格は？ (ラベル: 価格という数値)
    *   ある人の勉強時間や睡眠時間（これらが**特徴量 Feature** となる）から、その人のテストの点数（ラベル）を予測する。
*   **B) 分類 (Classification):** 予測したい**正解ラベル (Label)** が、**あらかじめ決められたいくつかのカテゴリ（種類）のうちのどれか**である場合。例えば…
    *   このメールは「迷惑メール」か「通常メール」か？ (ラベル: 「迷惑メール」「通常メール」の2つのカテゴリのいずれか)
    *   この画像に写っているのは「猫」か「犬」か「鳥」か？ (ラベル: 3つの動物カテゴリのいずれか)
    *   ある顧客が、過去の購買履歴や年齢、性別（これらが特徴量）といった情報から、次に新商品を買ってくれるか「買う」「買わない」のどちらかのカテゴリを予測する。

### **16.2.2 教師なし学習 (Unsupervised Learning)：お手本なしでデータ構造を探る**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/16-14.png)

*   **A) クラスタリング (Clustering):** 与えられたデータを、その**類似性に基づいていくつかのグループ（クラスター）に自動的に分ける**手法だ。どんなグループが形成されるかは、データの特徴からコンピューターが見つけ出す。
    *   例えば、たくさんの顧客の購買データを分析して、「お得情報に敏感な節約志向グループ」「新商品をいち早く試すトレンド追求グループ」「いつも同じ定番品を買い続ける保守的グループ」のように、顧客をいくつかのタイプに自動で分類する、といったことができる。
*   **B) 次元削減 (Dimensionality Reduction):** たくさんの特徴量（データの列、つまり次元）を持つ非常に複雑なデータを、その**本質的な情報をできるだけ失わずに、より少ない数の主要な特徴量（より低い次元）で表現し直す**手法だ。
    *   例えば、非常に多くのアンケート項目（何十個もの質問）があったとしても、実はそれらの回答パターンは「外向性」「協調性」「革新性」といった、いくつかの主要な潜在的な特性で大部分が説明できる、ということを見つけ出すようなイメージだね。データをよりシンプルに、そして理解しやすく（可視化しやすく）するのに役立つ。

### **データの概要：アイリスデータセットとは？**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/16-33.png)

```python
# コード例16-1

import pandas as pd
from sklearn.datasets import load_iris # アイリスデータセットを読み込むための関数

# アイリスデータセットを読み込む
iris_raw = load_iris()

# 読み込んだデータの中身を確認してみよう
print(type(iris_raw))
print(iris_raw.keys())
```

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/16-34.png)

*   `iris_raw.data`: これがメインのデータ、つまり**特徴量**の数値データだ。150個のアヤメのサンプルそれぞれについて、4つの特徴量（がく片の長さ `sepal length (cm)`、がく片の幅 `sepal width (cm)`、花びらの長さ `petal length (cm)`、花びらの幅 `petal width (cm)`）が記録されている。
*   `iris_raw.feature_names`: 上の `data` の各列が何を表すか、その**特徴量の名前**のリストだ。
*   `iris_raw.target`: これが**正解ラベル**のデータ。150個のサンプルそれぞれが、どの種類のアヤメなのかを**数値（0, 1, 2）で**示している。
*   `iris_raw.target_names`: 上の `target` の数値（0, 1, 2）が、それぞれ**どの花の名前**に対応するかのリストだ。具体的には、`0` が setosa (**ヒトツバアヤメ**)、`1` が versicolor (**ブルーフラッグ**)、`2` が virginica (**バージニアアイリス**) に対応している。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/16-35.png)

```python
# コード例16-2

# データフレームに変換
iris_df = pd.DataFrame(data=iris_raw.data, columns=iris_raw.feature_names)

# 列名を扱いやすく変更 (機械学習ではよく行われる)
iris_df.rename(
    columns={
        "sepal length (cm)": "sepal_length",
        "sepal width (cm)": "sepal_width",
        "petal length (cm)": "petal_length",
        "petal width (cm)": "petal_width",
    },
    inplace=True,
)

# ここまでの結果を表示して確認
iris_df.head()
```

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/16-36.png)

**`pd.Categorical.from_codes(codes, categories)`の使い方:**

*   **第一引数 (`codes`):** カテゴリの**番号（コード）**が入ったリストや配列を渡す。今回は `iris_raw.target` (`[0, 1, 2, 0, ...]`) がこれにあたる。
*   **第二引数 (`categories`):** その番号がそれぞれ何を意味するかの**カテゴリ名のリスト**を渡す。今回は `iris_raw.target_names` (`['setosa', 'versicolor', 'virginica']`) がこれにあたる。このリストの**順番が重要**で、第一引数の`codes`の`0`が`categories`の0番目、`1`が1番目…というように対応付けられるんだ。
*   **戻り値:** **`Categorical` オブジェクト**という、`pandas`の特別なデータ型を返す。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/16-37.png)

```python
# コード例16-3: `pd.Categorical.from_codes` を使って、数値ラベルをカテゴリ名に変換し、新しい列として追加
iris_df["species_name"] = pd.Categorical.from_codes(
    iris_raw.target, iris_raw.target_names
)

# 最終的なDataFrameを表示して確認
iris_df.head()
```

### **ステップ1：データの前処理**

**ア. 特徴量(X)とラベル(y)の準備**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/16-40.png)

*   **特徴量 X:** 将来的に複数の特徴量（例えば、花びらの長さと、がく片の長さ）を使うことを想定しているため、**2次元のデータ形式（DataFrameのような、行と列を持つ表の形）**で渡すのがお約束になっている。だから、たとえ特徴量が1つだけでも、`iris_df[['petal_length']]` のように角括弧を2重にして、2次元のDataFrameとして取り出すんだ。
*   **ラベル y:** 予測したい答えは常に1つの列だから、**1次元のデータ形式（Seriesのような、単なる値のリスト）**で渡すのが一般的だ。だから、`iris_df['petal_width']` と角括弧1つで取り出す。

**イ. 訓練データとテストデータへの分割**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/16-41.png)

**`train_test_split()`の主な引数と戻り値:**

*   **引数:**
    *   **最初の引数群:** 分割したいデータ（特徴量Xとラベルy）を順番に渡す。
    *   `test_size`: テストデータとして分割する割合を0から1の間の小数で指定する（例: `0.3`なら30%）。
    *   `random_state`: 分割の際のランダムな選び方を固定するための「シード値」。ここに同じ数値を指定すれば、何度実行しても必ず同じようにデータが分割されるので、分析の**再現性**を保つために非常に重要だ。
*   **戻り値:**

**`KMeans`の主なオプション（ハイパーパラメータ）:**

*   `n_clusters`: **作成したいクラスター（グループ）の数**を整数で指定する。これはk-means法で最も重要なパラメータで、事前に分析者が決める必要がある。
*   `n_init`: k-means法は、最初にランダムな点を配置することから始まるため、その初期値によって結果が少し変わることがある。このパラメータは、**異なる初期値で学習を何回実行するか**を指定する。`'auto'`に設定すると、scikit-learnが適切な回数を自動で判断してくれる。
*   `random_state`: k-means法で使われるランダム性を固定するためのシード値。これを指定すると、何度実行しても同じ結果が得られ、分析の再現性が保たれる。

1.  まず、指定された数（`k`）の「クラスターの中心」を、データ空間にランダムに配置する。
2.  **割当てステップ:** 全てのデータポイントを、最も距離が近い「クラスター中心」のグループに割り当てる。
3.  **更新ステップ:** 各グループに属するデータポイントの平均的な位置を計算し、その位置に「クラスター中心」を移動させる。
4.  この「割当て」と「更新」のステップを、「クラスターの中心」がほとんど動かなくなるまで繰り返す。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/16-125.png)

*   **`k`**: これは、我々分析者が「**データをいくつのクラスター（グループ）に分けたいか**」を**事前に指定する数**のことだ。
*   **`means`**: これは「**平均**」を意味する。まさに、アルゴリズムの「3. 更新ステップ」で各クラスターの中心を、そのグループに属するデータポイントの**平均的な位置**として計算していたね。

    *   この時点では、まだ**写真は撮られていない**。カメラマンが「よし、ここから撮るぞ！」と決めただけで、元のデータ（被写体）はまだ4次元のままだ。
2.  **変換: 実際にシャッターを押して「写真」を撮るプロセス。**
    *   見つけ出した主成分軸（カメラの向き）を使って、実際に元の4次元のデータを新しい2次元のスクリーンに**射影（投影）し、2次元のデータに変換する**。
    *   この変換作業を実行して初めて、我々が欲しい「2次元の写真（次元削減されたデータ）」が手に入るんだ。

```python
# コード例16-55

# 3. 学習と変換 (X_clf から主成分を学習し、同時にデータを2次元に変換)
X_iris_pca = pca_model.fit_transform(X_clf)
```

#### **b. 学習の準備：環境とハイパーパラメータの設定**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/16-159.png)

```python
# コード例16-61: 環境の定義
states = [0, 1, 2, 3]  # エージェントが取りうる全ての状態 (S, _, _, G)
actions = ["left", "right"]  # エージェントが取れる全ての行動
goal_state = 3  # ゴールの状態
```

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/16-160.png)

*   `alpha` (学習率): 新しい経験（報酬）を、どれくらい強くQテーブルに反映させるかの度合い (0～1)。
*   `gamma` (割引率): 目先の報酬だけでなく、将来得られる報酬をどれくらい重視するかの度合い (0～1)。
*   `episodes`: 試行錯誤を何回繰り返すか。

```python
# コード例16-62: ハイパーパラメータ
alpha = 0.1  # 学習率
gamma = 0.9  # 割引率
episodes = 1000  # 学習エピソード数
```

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/16-161.png)

```python
# コード例16-63

# QテーブルをDataFrameで表現してみる（初期状態）
q_table_df = pd.DataFrame(0.0, index=states, columns=actions)

# 学習前のQテーブル（カンニングペーパー）
q_table_df
```

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/16-162.png)

#### **c. 学習の実行：試行錯誤でQテーブルを更新**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/16-163.png)

```python
# コード例16-64: 学習ループ
import random

for episode in range(episodes):
    # --- 1. エピソードの開始 ---
    current_state = 0  # 各エピソードの開始時に、エージェントをスタート地点に戻す

    # --- 2. ゴールするまで行動を繰り返す ---
    while current_state != goal_state:
        # --- 2a. 行動の選択 ---
        action = random.choice(
            actions
        )  # 今はランダムに行動を選択して、迷路を「探索」する

        # --- 2b. 環境のシミュレーション ---
        # 選んだ行動によって、次の状態がどうなるか計算
        if action == "left":
            next_state = current_state - 1
        else:  # 'right'
            next_state = current_state + 1

        # 次の状態に応じて報酬を決定
        if next_state == goal_state:
            reward = 10  # ゴールに到達！
        elif next_state < 0 or next_state >= len(states):  # 壁に衝突
            reward = -1
            next_state = current_state  # 状態は変化させない
        else:
            reward = 0  # 何もないマス

        # --- 2c. Q値の更新 (学習の核心！) ---
        # 更新前のQ値を取得
        old_q_value = q_table_df.loc[current_state, action]

        # 次の状態で得られる最大のQ値（将来価値）を取得
        # もし次がゴールなら、その先に価値はないので0とする
        next_max_q = q_table_df.loc[next_state].max() if next_state != goal_state else 0

        # Q学習の更新式
        new_q_value = (1 - alpha) * old_q_value + alpha * (reward + gamma * next_max_q)

        # 計算した新しいQ値でQテーブルを更新
        q_table_df.loc[current_state, action] = new_q_value

        # --- 2d. 状態の更新 ---
        current_state = next_state  # エージェントを次の状態に移動させる

print("学習が完了しました！")
```

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/16-164.png)

*   **`for episode in range(episodes):`**: これは外側のループで、「スタートからゴールまで」という一連の挑戦（**エピソード**）を1000回繰り返すためのものだ。
*   **`current_state = 0`**: 各エピソードの開始時には、必ずエージェントをスタート地点（状態`0`）に戻す。
*   **`while current_state != goal_state:`**: これは内側のループで、エージェントがゴールにたどり着くまで、一つのエピソード内での行動を続けるためのものだ。
*   **`action = random.choice(actions)`**: ここが「試行錯誤」の心臓部だ。学習の初期段階では、エージェントはまだ賢くないので、**ランダムに**次の行動（`'left'`か`'right'`）を選んで、色々な道を試してみる（**探索**）。
*   **状態遷移と報酬の決定**: 選んだ行動によって、次の状態（`next_state`）と、その結果得られる報酬（`reward`）が、我々が設定したルールに従って（Q学習の更新式）計算される。
*   **Q値の更新**: そして、ここがQ学習の最も重要な計算部分だ。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/16-165.png)

*   **`old_q_value`**: これは、更新前のQテーブルに書かれていた、今の「状態」と「行動」に対する「古い評価点」。
*   **`reward + gamma * next_max_q`**: これが、今回の行動で得られた「**新しい知見**」だ。
    *   `reward`: 今回の行動で**すぐにもらえた報酬**（目先の利益）。
    *   `next_max_q`: 行動した**次の状態で、取りうる最善の手（Q値が最大になる手）の価値**。つまり、「**将来期待できる利益**」だ。`gamma`（割引率）を掛けることで、少しだけ割り引いて考える。
*   **`(1 - alpha) * ... + alpha * ...`**: 最後に、学習率 `alpha` を使って、「古い評価点」と「新しい知見」を**重み付けして平均**している。「学習率`alpha`の分だけ、新しい知見を信じて更新しよう」という調整だね。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/16-166.png)

```python
# コード例16-68: 学習した方策（ポリシー）でゴールを目指してみる

# --- 1. 初期化 ---
current_state = 0  # スタート地点から開始
path = [current_state]  # たどった経路を記録するリスト

# --- 2. ゴールするまで最適な行動を繰り返す ---
while current_state != goal_state:
    # --- 2a. 最適な行動の選択 (探索ではなく活用！) ---
    # q_tableから現在の状態に対応する行を取り出し、その中でQ値が最大になる列名(行動)を取得
    action = q_table_df.loc[current_state].idxmax()

    # --- 2b. 選択した行動の表示 ---
    print(
        f"状態 {current_state} で {action} を選択 (Q値: {q_table_df.loc[current_state, action]:.2f})"
    )

    # --- 2c. 状態の更新 ---
    # 選択した行動に従って、次の状態に移動
    if action == "left":
        current_state -= 1
    else:  # 'right'
        current_state += 1

    # --- 2d. 経路の記録 ---
    path.append(current_state)

# --- 3. 結果の表示 ---
print(f"\nゴール！ Path: {path}")
```

*   **機械学習の定義と基本:** コンピューターがデータ（経験）から学習し、パターンを発見し、予測や判断を行う技術。
*   **用語:** 「ラベル」（予測対象/被説明変数）と「特徴量」（手がかり/説明変数）。
*   **学習スタイルによる主な種類:**
    *   **教師あり学習:** 正解ラベル付きデータで学習（**回帰**で数値を予測、**分類**でカテゴリを判別）。
    *   **教師なし学習:** ラベルなしデータから構造発見（**クラスタリング**でグループ分け、**次元削減**で情報を要約）。
    *   **強化学習:** 環境との相互作用と報酬/罰で最適行動を学習。
*   **機械学習の実践手順の概要:**
    1.  **データの前処理:** データクリーニング、特徴量のスケーリング（標準化など）、データの分割（訓練データとテストデータ、ホールドアウト法や交差検証）。
    2.  **モデルの種類の選択:** 課題の種類（回帰、分類、クラスタリングなど）に応じて、適切なモデルの仮定や方向性を選択。
    3.  **学習 (Training / Fitting):** 訓練データを使って、モデルのパラメータをアルゴリズムに従って最適化。
    4.  **評価 (Evaluation):** テストデータを使って、学習済みモデルの汎化性能を評価。回帰ではMSEなど、分類では正解率、適合率、再現率、F1スコアなど。
*   **教師あり学習のモデル例 (`scikit-learn` を使用):**
    *   **回帰:** `LinearRegression`（線形回帰）。
    *   **分類:** `DecisionTreeClassifier`（決定木、木構造の可視化による解釈性の高さ）、`RandomForestClassifier`（ランダムフォレスト、アンサンブル学習による性能向上）、`MLPClassifier`（ニューラルネットワークの基礎）。
*   **教師なし学習のモデル例 (`scikit-learn` を使用):**
    *   **クラスタリング:** `KMeans`（k-means法）。
    *   **次元削減:** `PCA`（主成分分析）。
*   **強化学習の例:** Q学習の基本的な考え方と、簡単なPythonコードでの実装。
*   **深層学習 (ディープラーニング):** ニューラルネットワークを深く用いる強力な手法であり、様々な学習スタイルと組み合わせて現代AIの中核を成していること。
*   **機械学習の注意点:** データの質と量、過学習、モデルの解釈性と精度のバランス、バイアスと公平性。
