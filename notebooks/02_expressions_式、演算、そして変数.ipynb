さて、四則演算以外にも、Pythonには便利な計算機能がある。

- **べき乗** ( `**` )： 同じ数を何度も掛け合わせる計算。例えば `2 ** 3` は `2` を 3 回掛ける (`2` × `2` × `2`) という意味だ。
- **整数除算** (`//`)： 割り算の答えの「整数部分だけ」を求める。例えば `7 // 3` は `2` になる。
- **剰余** (`%`)： 割り算の「余り」を求める。例えば `7 % 3` は `1` になる。

```python
# コード例2-5: べき乗 (2の3乗)
2**3
```

```python
# コード例2-6: 整数除算 (7割る3の商)
7 // 3
```

```python
# コード例2-7: 剰余 (7割る3の余り)
7 % 3
```

### **2.2.6 自分自身への代入（値を更新する）**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/2-27.png)


```python
# コード例2-22: count に 0 を代入
count = 0
print(count)
```

```python
# コード例2-23: count の現在の値 (0) に 1 を足して、結果を再び count に代入
count = count + 1
print(count)
```

```python
# コード例2-24: もう一度 count を 1 増やす
count = count + 1
print(count)
```

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/2-28.png)


`count = count + 1` という行は、コンピューターの中では次のようなステップで実行されると考えてみてほしい。

1. **右辺の評価:** まず、`=` の右側にある `count + 1` を計算する。
   - この時点での `count` の値（例えば最初なら `0`）を箱から取り出す。
   - 取り出した値 `0` に `1` を足して、`1` という計算結果を得る。
2. **左辺への代入:** 次に、ステップ1で得られた計算結果 `1` を、`=` の左側にある変数 `count` の箱に**代入（上書き）**する。
   - これで、`count` の箱の中身が `0` から `1` に更新される。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/2-29.png)


![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/2-30.png)


### **2.2.7 複合代入演算子（累積代入）**

変数 = 変数 + 値 のような、自分自身の値を更新する書き方はよく使われるので、もっと短く書くための **複合代入演算子 (augmented assignment operator)** が用意されている。

- `+=` : `x += 1` は `x = x + 1` と同じ
- `-=` : `x -= 1` は `x = x - 1` と同じ
- `*=` : `x *= 2` は `x = x * 2` と同じ
- `/=` : `x /= 2` は `x = x / 2` と同じ
- `%=` : `x %= 3` は `x = x % 3` と同じ
- `//=` : `x //= 3` は `x = x // 3` と同じ
- `**=` : `x **= 2` は `x = x ** 2` と同じ

```python
# コード例2-25: score に 100 を代入
score = 100
print(score)
```

```python
# コード例2-26: score に 10 を加える
score += 10  # score = score + 10 と同じ
print(score)
```

```python
# コード例2-27: score を 2 倍にする
score *= 2  # score = score * 2 と同じ
print(score)
```

## **2.3 データ型：数字？文字？データの種類を見極めよう！**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/2-35.png)


**基本的なデータ型**

- **int (整数型):** 小数点のない数。例: 10, 0, -5
- **float (浮動小数点数型):** 小数点のある数。例: 3.14, 10.0, -0.5
- **str (文字列型):** 文字の並び。シングルクォート (`'`) またはダブルクォート (`"`) で囲む。例: `'こんにちは'`, `"Python"`, `'123'` (数字でもクォートで囲めば文字列)
- **bool (ブール型):** 真偽値。`True` (真) または `False` (偽) のどちらかの値しか取らない。条件分岐などで使う。 (これは後の章で詳しく学ぶよ)

## **2.4 今回の冒険のまとめ**

この章では、次のことを学びました。

*   **式と演算子:**
    *   **主な演算子:**
        *   四則演算: `+` (足し算), `-` (引き算), `*` (掛け算), `/` (割り算)
        *   その他: `**` (べき乗), `//` (整数除算), `%` (剰余)
    *   **優先順位:** `**` が最も高く、次に `*`, `/`, `//`, `%`、最後に `+`, `-` の順で計算される。括弧 `()` を使うと優先順位を変更できる。
*   **変数:**
    *   **役割:** データ（数値や文字など）に名前をつけて一時的に保存するための仕組み。
    *   **代入:** `=` 演算子を使用。右辺の値を評価し、左辺の変数に格納（上書き）する。
    *   **参照:** 変数名を使って、格納されている値を読み出す。
    *   **命名規則:**
        *   使用可能な文字: 半角の英字 (`a-z`, `A-Z`)、数字 (`0-9`)、アンダースコア (`_`)。
        *   開始文字: 英字またはアンダースコアでなければならない（数字で始まらない）。
        *   予約語: `if`, `for`, `while` など、Pythonが特別な意味を持つ単語は使用不可。
        *   大文字/小文字: 区別される (`price` と `Price` は別の変数)。
    *   **アンパック代入:** `x, y = 10, 20` のように、複数の変数に一度に値を代入できる（左右の要素数は一致させる）。
    *   **自己代入:** `count = count + 1` のように、変数自身の現在の値を使って計算し、結果を同じ変数に再代入する操作。
    *   **複合代入演算子:** `+=`, `-=`, `*=` など、自己代入を短縮して記述する演算子。
    *   **`input()` 関数:** プログラム実行中にユーザーからのキーボード入力を受け付ける関数。（**注意点：** 受け取った値は常に文字列 (`str`) 型になる。）
*   **データ型:**
    *   **データ型:** データの種類を示す分類。
        *   `int`: 整数 (例: `10`, `-5`, `0`)
        *   `float`: 浮動小数点数 (例: `3.14`, `-0.5`, `10.0`)
        *   `str`: 文字列 (例: `"こんにちは"`, `'Python'`, `'123'`)
        *   `bool`: 真偽値 (`True` または `False`)
    *   **リテラル:** プログラム中に直接記述された値のこと (例: `100`, `3.14`, `"Hello"`)。
    *   **`type()` 関数:** 変数やリテラルのデータ型を調べるための関数。 (例: `type(10)` は `<class 'int'>` を返す)
