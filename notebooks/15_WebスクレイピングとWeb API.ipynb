### **15.2.2 Webスクレイピングの基本的な流れ**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/15-27.png)

**インターネット通信の基本：リクエストとレスポンス**

1.  **リクエスト (Request / お願い):**
    *   君のパソコンやスマホ（これを**クライアント Client** と呼ぶ）に入っているブラウザが、指定されたURLの情報を管理しているコンピューター（**サーバー**）に対して、「**このWebページの情報を見せてください！**」とお願いのメッセージを送るんだ。これが**リクエスト**だ。
2.  **レスポンス (Response / お返事):**
    *   リクエストを受け取ったウェブサーバーは、そのお願いに応じて、Webページを構成するデータ（主に**HTMLファイル**という、ページの構造が書かれた設計図のようなもの）を、君のクライアントに送り返してくれる。これが**レスポンス**だ。
3.  **ブラウザによる表示:**
    *   クライアントのブラウザは、受け取ったHTMLファイルなどの情報を解読（**解析 Parse**）し、我々が見慣れたウェブページの形に組み立てて画面に表示する。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/15-28.png)

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/15-29.png)

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/15-30.png)

1.  **リクエスト (Request) の送信とレスポンス (Response) の受信:** 我々のPythonプログラムが、「このWebページの情報が欲しい！」と対象Webサイトのサーバーにお願いを送り、サーバーからのWebページのデータ（主にHTML）を受け取る。
2.  **解析 (Parse):** Pythonプログラムが、受け取ったHTMLを解析し、扱いやすい形に構造化する。
3.  **情報特定**：ブラウザの「開発者ツール」を使い、取得したい情報が解析対象のHTMLドキュメント内のどこに位置しているのかを正確に特定する。
4.  **抽出:** 解析されたHTMLの中から、指定した条件（開発者ツールで調べたタグや属性など）に合う部分を探し出し、必要な情報を抜き出す。
5.  **保存・活用:** 抜き出した情報を、ファイルに保存したり、DataFrameに入れたりして活用する。

### **15.2.4 スクレイピングの道具：`requests`、`BeautifulSoup`、そして `time`**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/15-39.png)

1.  **`requests` (リクエスト): 魔法の「設計図取り寄せの呪文」**
    *   特定の建物の住所（URL）を指定すると、その建物の基本設計図（主にHTMLコード）を、遠く離れた島（サーバー）から**ごっそり取り寄せてきてくれる**ライブラリだ。まずはこれで、お宝が眠る遺跡の設計図を手に入れる。
2.  **`BeautifulSoup` (ビューティフルスープ): 古代文字解読のエキスパート**
    *   `requests` が取り寄せてくれた設計図（HTMLコード）は、そのままでは複雑で読みにくい古代文字の羅列のようなものだ。`BeautifulSoup` は、この設計図を解析し、建物の各部屋や通路、壁画の場所（タグや属性）といった構造ごとに綺麗に整理し、**Pythonプログラムが「ここが宝物庫だ」「ここに碑文がある」と理解しやすい形に「解読」してくれる**ライブラリだ。これによって、特定のお宝（情報）が設計図のどこにあるかを見つけ出し、取り出すことができるようになる。
3.  **`time` (タイム): 遺跡への敬意を示す「探査インターバルタイマー」**
    *   そして、もう一つ忘れてはならないのが、この `time` モジュールだ。これはPythonの標準ライブラリの一つで、プログラムの探査活動を一時的に停止させる `sleep()` という機能を持っている。Webサイトのサーバーに対して短時間で何度も探査ロボットを送り込むと、サーバーに大きな負荷をかける恐れがある。これは絶対に避けなければならないマナー違反であり、場合によってはアクセスを拒否される原因にもなる。この `time.sleep()` を使って、**設計図の取り寄せ依頼（リクエスト）と次の依頼の間に適切な「休憩時間」を入れる**ことで、サーバーに優しく、敬意を払ったリクエストができるようになるんだ。

##### **ステップ1: リクエストの送信とレスポンスの受信！ 建物の設計図（HTML）の入手**

**`requests.get()` の基本的な使い方:** `requests.get('取得したいWebページのURL', timeout=秒数)`

*   最初の引数には情報を取得したいWebページの**URL**を文字列で指定する。
*   `timeout` 引数（オプションだが重要！）: サーバーからの応答が指定した秒数以内に返ってこない場合、処理を打ち切ってエラーを出すように設定できる。インターネットの状況によっては応答が遅いこともあるから、むやみに待ち続けないための安全装置だね。今回は10秒に設定してみよう。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/15-45.png)

````python
# コード例15-3

# 探査対象の建物の住所 (ターゲットURL)
url = "https://books.toscrape.com/index.html"

# requests.get() で建物の設計図 (HTML) を取得し、responseオブジェクトに格納
response = requests.get(url, timeout=10)
````

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/15-46.png)

**`response.status_code` とは？**
これは、サーバーからの応答がどんな状態だったかを示す「**ステータスコード**」という3桁の数字だ。たくさんの種類があるが、特に覚えておきたいのは…

*   **`200` (OK):** リクエスト成功！ サーバーは問題なく情報を提供してくれた。一番嬉しいお返事だね。
*   `404` (Not Found): 指定したページが見つからなかった…。「お探しの本はありませんでした」という感じ。
*   `403` (Forbidden): アクセスが禁止されている。「この本は貸し出せません！」と言われたようなもの。
*   `500` (Internal ServerError): サーバー側で何か問題が発生した。「図書館が臨時休館中です…」みたいな状況。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/15-47.png)

````python
# コード例15-4: サーバーからの応答状態 (ステータスコード) を確認
print(f"ステータスコード: {response.status_code}")
````

#### **ステップ2: 解析！ `BeautifulSoup` で設計図（HTML）を解読する**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/15-50.png)

**`BeautifulSoup` の基本的な使い方:** `BeautifulSoup(解析したいHTML文字列, '使用するパーサーの種類')`

*   最初の引数には、`requests.get().text` で取得したHTMLの文字列（今回の場合は `html_content` 変数）を渡す。
*   2番目の引数には、「**HTMLパーサー (HTML Parser)**」の種類を文字列として指定する。これは「どの言語の専門家（解読器）に解読をお願いするか」というようなものだね。今回は手軽な `'html.parser'` を使おう。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/15-51.png)

*   **`'html.parser'`**: Pythonの標準ライブラリに含まれているパーサー。追加のインストールが不要で手軽に使える。多くの場合、これで十分だ。
*   `'lxml'` や `'html5lib'`: より高速だったり、崩れたHTMLに対してもより頑健だったりする外部ライブラリのパーサー（別途 `pip install lxml` や `pip install html5lib` が必要）。

**A. 複数の「お宝の原石」をまとめて発見する：`soup.find_all()`**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/15-57.png)

**BeautifulSoupオブジェクト`.find_all()`メソッドの基本的な使い方**: `BeautifulSoupオブジェクト.find_all(タグ名, attrs={'属性名': '属性値'})`
`
- **実行するオブジェクト:** `BeautifulSoup` でHTML全体を解析した直後の、大きな `soup` オブジェクト。
- **最初の引数**：探したい**HTMLタグ名**（例: `'article'`, `'h3'`, `'p'`）を文字列で指定する。
- オプションとして、その後ろにそのタグが持つべき**属性とその値**を `属性名='属性値'` の形で指定できる。特にHTMLの `class` 属性を指定する場合は、Pythonの予約語 `class` と区別するため、**`class_`** のようにアンダースコアを付けて `class_='クラス名'` と書くのがお約束だ。
- **返ってくるもの:** 条件に一致したHTML要素が、それぞれ**個別の `BeautifulSoup` の要素オブジェクトとして格納されたリスト**だ。何も見つからなければ空のリストが返る。

````python
# コード例15-8

# 各本に対応する <article class="product_pod"> 要素を全て取得
book_articles = soup.find_all("article", class_="product_pod")

# 見つかった本の数を表示してみる
print(f"見つかった本の数: {len(book_articles)}")
````

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/15-58.png)

**B. 個別の区画（要素オブジェクト）から特定のお宝を掘り出す：要素オブジェクト`.find()` と、入れ子構造の探索**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/15-59.png)

**要素オブジェクト`.find()` メソッドの基本的な使い方:** `要素オブジェクト.find(タグ名, 属性名1=属性値1, ...)`

*   **実行するオブジェクト:** `find_all()` などで見つけ出された、**個別のHTML要素を表す `BeautifulSoup` の要素オブジェクト**（例えば、ループで取り出した `article` 変数）。
*   引数の指定方法は `soup.find_all()` と同様。
*   **返ってくるもの:** その要素オブジェクトの**内側**で、条件に一致した**最初の1つのHTML要素オブジェクト**。もし何も見つからなければ `None` という特別な値が返る。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/15-60.png)

**C. お宝の「中身」を取り出す：`.text` と 属性アクセス `要素['属性名']`**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/15-61.png)

*   タグに囲まれた**文字情報**（例えば、価格の数字）を取得したい場合は、見つけた要素オブジェクトの **`.text`** という属性を参照する。
*   タグが持っている**属性の値**（例えば、本のタイトルが書かれた`title`属性の値）を取得したい場合は、辞書のように**角括弧 `[]` と属性名**を使って `要素['属性名']` のようにアクセスする。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/15-62.png)

*   **タイトルのお宝:** `article`要素オブジェクト の中の `<h3>` タグ要素オブジェクトを探し、さらにその中の `<a>` タグ要素オブジェクトを探し、その `<a>` タグ要素オブジェクトが持つ `title` 属性の値を取得する。
*   **価値のお宝:** `article`要素オブジェクト の中の `<p class="price_color">` タグ要素オブジェクトを探し、その中のテキスト（価値を示す文字列）を取得する。

````python
# コード例15-9

# 抽出した情報を格納するための空のリストを準備
book_titles = []
book_prices = []

# 各 <article> 要素 (各本) についてループ処理
for article in book_articles:
    # --- タイトルの抽出 ---
    h3_tag = article.find("h3")
    a_tag = h3_tag.find("a")
    title = a_tag["title"]
    book_titles.append(title)

    # --- 価格の抽出 ---
    price_tag = article.find("p", class_="price_color")
    price_text = price_tag.text
    price_float = float(price_text.replace("£", ""))
    book_prices.append(price_float)

# 抽出した最初の5件のタイトルと価格を表示して確認
for i in range(min(5, len(book_titles))):
    print(f"タイトル: {book_titles[i]}, 価格: {book_prices[i]}")
````

### **15.3.1 Web APIの仕組みとメリット**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/15-69.png)

**Web APIの主なメリット (スクレイピングとの比較):**

1.  **公式性・安定性:**
    *   APIはサービス提供者によって公式に設計・提供・維持されているため、WebページのHTML構造のように頻繁に変わることが少なく、比較的**安定して**データを取得し続けることができる。サイトのデザイン変更の影響を受けにくい。
2.  **合法性:**
    *   APIには通常、明確な**利用規約 (Terms of Service)** が定められている。この規約に従って利用する限り、著作権侵害や不正アクセスといった**法的リスクを大幅に低減**できる。何が許され、何が禁止されているかがハッキリしている。
3.  **効率的なデータ形式:**
    *   多くの場合、APIは必要なデータだけを、プログラムで扱いやすい**構造化された形式（例えばJSON形式やXML形式）**で直接返してくれる。HTMLの中から情報を探し出して解析する手間が省け、より**効率的**で**正確な**データが得られる。
4.  **アクセス制御と負荷軽減:**
    *   APIはプログラムからの利用を前提としているため、多くの場合、認証のための**APIキー**の使用や、**利用回数・データ取得量の上限（レートリミット）**などが設定されている。これにより、サーバーへの過度な負荷を防ぎ、サービス全体の安定性を保つ仕組みになっている。節度ある利用が促される。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/15-70.png)

*   **JSON (JavaScript Object Notation):** `{"お宝の名前": "賢者の石"}` のような `キー: 値` のペア（Pythonの辞書に似ているね！）や、値のリスト（Pythonのリストに似ている）を組み合わせてデータを表現する、非常に軽量で分かりやすい形式。最近のWeb APIでは主流だ。
*   **XML (eXtensible Markup Language):** HTMLのようにタグを使ってデータを階層的に表現する形式。JSONより少し複雑だが、古くから使われている。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/15-71.png)

3.  検索結果がたくさん出てくると思うが、その中から、我々の目的に合致しそうな統計表を探す。今回は、国勢調査の一部として提供されている**「最終報告書「日本の人口・世帯」統計表」**の中にあるものが良さそうだ。
4.  さらに絞り込んでいくと、「**人口及び人口の指数－全国，都道府県（大正９年～平成27年）**」という名前の統計表が見つかるはずだ。これこそが我々の探している巻物だ！
5.  この統計表の紹介ページ（[URL](https://www.e-stat.go.jp/stat-search/database?page=1&layout=datalist&toukei=00200521&tstat=000001080615&cycle=0&tclass1=000001124175&statdisp_id=0003411191&result_page=1&tclass2val=0)）の情報の中に「**統計表表示 ID**」という項目がある。見てごらん、そこに `0003411191` と書かれているだろう？ これが、この巻物の固有番号、つまり我々が必要な `statsDataId` なんだ。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/15-98.png)

````python
# コード例15-16

# 統計データをJSON形式で取得するAPIのエンドポイント (バージョン3.0)
base_url = "https://api.e-stat.go.jp/rest/3.0/app/json/getStatsData"

# e-Statで確認した2015年国勢調査の都道府県別総人口が含まれる統計表のID
target_stats_data_id = "0003411191"

# 組み立てるリクエストURL (appIdとstatsDataIdは必須)
request_url = f"{base_url}?appId={MY_APP_ID}&statsDataId={target_stats_data_id}"
print(f"組み立てたリクエストURL: {request_url}")
````

````python
# コード例15-21

# 地域コード → 地域名マッピングを抽出
area_class = data["GET_STATS_DATA"]["STATISTICAL_DATA"]["CLASS_INF"]["CLASS_OBJ"]

# 最初の3件だけ表示
for item in area_class[:3]:
    print(f"データ: {item}")
````

````python
# コード例15-22

area_mapping = {}
time_mapping = {}

for obj in area_class:
    if obj["@id"] == "area":
        area_mapping = {item["@code"]: item["@name"] for item in obj["CLASS"]}
    elif obj["@id"] == "time":
        time_mapping = {item["@code"]: item["@name"] for item in obj["CLASS"]}

# DataFrameに変換
records = []
for item in values:
    value = item["$"]
    if item["@tab"] != "020":
        continue
    area_code = item["@area"]
    time_code = item["@time"]
    records.append(
        {
            "地域": area_mapping.get(area_code, area_code),
            "年": time_mapping.get(time_code, time_code),
            "人口": None if "-" in value else int(value),
        }
    )

df = pd.DataFrame(records)

# 表示
df
````

*   **データIDの確認方法:** 世界銀行の公式データサイト (例: https://data.worldbank.org/ ) をブラウザで開いて、興味のある指標を探してみよう。指標を見つけたら、その詳細ページに進み、「Details」ボタンなどをクリックすると、その指標の固有IDが表示されているはずだ。例えば…
    *   「[一人当たりGDP（現在の米ドル価格）](https://data.worldbank.org/indicator/NY.GDP.PCAP.CD?view=chart)」のIDは `NY.GDP.PCAP.CD` だったりする。
    *   「[一人当たり二酸化炭素排出量（土地利用や林業による影響を除いた数値）](https://data.worldbank.org/indicator/EN.GHG.CO2.PC.CE.AR5)」のIDは `EN.GHG.CO2.PC.CE.AR5` だったりする。
*   **国コードの確認方法:** 通常、国際標準化機構 (ISO) が定める2文字または3文字のアルファベット国コード (ISO 3166-1 alpha-2 / alpha-3) が使われる。
    *   例: 日本は `JP` または `JPN`、アメリカは `US` または `USA`、中国は `CN` または `CHN`。
    *   `wb.get_countries()` というのを実行すると、`pandas_datareader` が認識している国名とそのISOコードの一覧をDataFrameとして取得できる。

この章では、次のことを学びました。

*   **Webからのデータ収集の意義:** ビッグデータ時代において、自動収集は不可欠。APIがあればAPIを優先。スクレイピングは最終手段で細心の注意を払い、倫理的に。
*   **Webスクレイピング:**
    *   Webページから情報を自動抽出。
    *   **Webページの基本構造:** HTML (骨格)、CSS (見た目)、JavaScript (動き)。サーバー、ドメイン、ファイルの関係。開発者ツールでHTML確認。
    *   **最重要注意点:** 著作権法、サイト利用規約 (`robots.txt`)、サーバー負荷への配慮を**絶対に遵守**。違法・迷惑行為は厳禁！
    *   **道具:** `requests` (HTML取得)、`BeautifulSoup` (HTML解析)。
    *   **実践 (練習サイト `books.toscrape.com`):** HTML取得 → 解析 → ターゲット特定 → 情報抽出 (`find_all`, `.text`等) → DataFrame化。
    *   サイト側の対策の存在も認識。
*   **Web API (推奨される方法):**
    *   サービス提供者が公式に用意したデータ取得窓口。安定・安全・明確。
    *   **利用の流れ:** ドキュメント熟読 → (必要なら)APIキー取得 → `requests.get()` でリクエスト (URL, パラメータ) → `.json()` でレスポンス解析 → データ活用。
    *   **実践例:** 郵便番号検索API (zipcloud)。
    *   **発展例:** e-Stat API (政府統計) の紹介。
*   **`pandas_datareader` の活用:**
    *   様々なオンラインデータソース (世界銀行, FRED, Stooqなど) から簡単にデータをDataFrameとして取得できるライブラリ。
    *   世界銀行データの取得 (`wb.download()`) と可視化。
    *   Stooq株価データの取得 (`web.DataReader()`) と可視化（`mplfinance` でローソク足チャート描画）。
    *   FRED経済指標データの取得 (`web.DataReader()`) と可視化。
