### **14.3.1 モデルの表現：`formula` で関係性を記述する**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/14-13.png)

$Y_i = \alpha + \beta_1 X_{1i} + \beta_2 X_{2i} + \dots + \varepsilon_i$

*   $Y_i$: **被説明変数**（結果となる変数、例：消費）
*   $X_{1i}, X_{2i}, \dots$: **説明変数**（原因となる変数、例：所得、人口）
*   $\alpha$: **切片**（すべての説明変数が0の時の$Y_i$の値、グラフの$Y$軸との交点）
*   $\beta_1, \beta_2, \dots$: **回帰係数**（各説明変数が$Y_i$にどれくらい影響を与えるかの度合い）
*   $\varepsilon_i$: **誤差項**（モデルでは説明しきれない、その他の要因によるズレ）。我々のモデルでは直接扱えない「ズレ」の部分だね。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/14-14.png)

回帰モデル：$消費_i = \alpha + \beta 所得_{i} + \varepsilon_i$

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/14-15.png)

```python
# コード例14-5: 消費(Y) を 所得(X1) で説明するモデルのformula
my_formula = "消費 ~ 1 + 所得"
```

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/14-16.png)

**`statsmodels` の `formula` の基本ルール:** `被説明変数名 ~ 1 + 説明変数名1 + 説明変数名2 + ...`

*   `~` (チルダ): 「は〜によって説明される」という意味。
*   `+`: 説明する要素を繋げる記号。

*   **切片 (定数項 $\alpha$) の扱い:** 
    *   `formula` の右辺に **`1`** を書くと、「**切片（定数項）を含めてくださいね**」という明示的な指示になる。
    *   実は、`statsmodels.formula.api` の `ols` 関数では、デフォルトで**自動的に切片が含まれる**ようになっている。だから、`'消費 ~ 所得'` と書くだけでも、裏では `'消費 ~ 1 + 所得'` と同じように切片を考慮したモデルが作られるんだ。
    *   もし、何らかの理由で**切片を含めたくない**（原点を通るモデルを考えたいなど、特殊な場合）ときは、`'消費 ~ 所得 - 1'` のように、**`- 1`** を明示的に書くことで切片を除外できる。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/14-17.png)

### **14.3.2 モデルの設定と推定：OLS（最小二乗法）の魔法で最適な線を見つける！**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/14-18.png)

**ステップ1：モデルの設定 (`smf.ols`)**

`smf.ols()` 関数の基本的な使い方：`smf.ols(formula='Y ~ X1 + X2', data=使用するDataFrame)`

*   `formula` 引数: ここに、分析したい変数間の関係性を記述した**モデルの設計図**（**formula文字列**）を指定する。「この設計図に基づいて分析の準備をしてね！」という意味だ。
*   `data` 引数: ここに、実際に分析に使用するデータが入った **`pandas` の DataFrameオブジェクト**を指定する。「このデータを使ってね！」という意味だ。ただし、`formula` 文字列の中で使われている変数名（被説明変数や説明変数）は、この `data` 引数で指定したDataFrameの中に、**同じ名前の列として存在している必要がある**。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/14-19.png)

*   `formula` には、さっき作った `'消費 ~ 1 + 所得'` という**モデルの設計図**が入っている変数 **`my_formula`** を渡す。
*   `data` には、消費と所得のデータが入っているDataFrame **`df`** を渡す。

```python
# コード例14-6
# 回帰モデル（所得だけで消費を説明）の設定
model = smf.ols(formula=my_formula, data=df)

# 設定済みのモデルオブジェクトを表示
model
```

### **14.4.1 仮説検定の基本的な考え方（超入門）**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/14-28.png)

**【仮説検定の3ステップまとめ】**

1. **わざと逆の仮定をする（帰無仮説）**：「本当はまったく関係ない（効果ゼロ）」と仮定する。
2. **偶然起きる確率を計算する**：「関係ない」とした場合、今回のデータのような結果が偶然出る確率を計算する。
3. **確率が低ければ論破！（棄却）**：その確率があまりにも低ければ「偶然とは考えにくい」と判断し、「関係ない」という仮定を捨てて、「関係がある！（対立仮説）」と結論づける。

### **14.4.3 より頑健な推定：不均一分散という落とし穴と、ロバスト標準誤差という名の盾！**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/14-42.png)

**`cov_type` で指定できる主なロバスト標準誤差の種類:** `statsmodels` では、いくつか異なる計算方法のロバスト標準誤差が用意されているが、よく使われるのは、White (1980) によって提案された方法や、それに基づくいくつかの修正版だ。これらは通常、`'HC'` (Heteroskedasticity-Consistent の略) という接頭辞で呼ばれる。

*   **`'nonrobust'` (デフォルト):** これが通常の、均一分散を仮定した標準誤差。
*   **`'HC0'`:** Whiteのオリジナルな不均一分散一致標準誤差。
*   **`'HC1'`:** MacKinnon and White (1985) による、有限標本（サンプルサイズが小さい場合）での性能を改善するための修正版。多くの統計ソフト（例えばStataなど）でデフォルトとして採用されていることが多い。**迷ったら、まずはこれを使ってみるのが良いだろう。**
*   **`'HC2'`:** さらに別の修正版で、より不偏性（推定値が真の値から偏らない性質）を重視したもの。
*   **`'HC3'`:** ジャックナイフ法と呼ばれる手法に基づいた修正版で、サンプルサイズが非常に小さい場合に特に推奨されることがある。一般的に、`HC1`よりもさらに保守的（標準誤差が大きめに出やすい＝有意になりにくい）な結果になる傾向がある。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/14-43.png)

```python
# コード例14-10: 回帰分析の実行。.fit() の cov_type 引数に 'HC1' を指定！
robust_results = model.fit(cov_type="HC1")

# 詳細な推定結果を表示
print(robust_results.summary())
```

### **14.5.3 秘密兵器 `regmonkey.regress` 関数の登場！**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/14-47.png)

**`regmonkey.regress()` 関数の基本的な使い方と主な引数:**

`regress(variables_dicts, df, decimal_places=2, lang="ja", cov_type="HC1")`

*   **`variables_dicts` (最初の引数・必須):** 分析したいモデルの情報を「辞書」の形で作り、それをさらに **「リスト」** に入れたものを渡すんだ。複数のモデルを一度に分析できるのが大きな特徴だね。各モデルの情報（辞書）は、例えば日本語なら `{'被説明変数': 'Y列名', '説明変数': ['X1列名', 'X2列名', ...]}` のように書く。英語なら `{'y': 'Y列名', 'X': ['X1列名', ...]}`。
*   **`df` (2番目の引数・必須):** 分析に使用する **`pandas` の DataFrame** を指定する。
*   `decimal_places` (オプション): 結果の表に表示される数値の、**小数点以下の桁数**を指定する。デフォルトは `2`（小数点以下2桁）。
*   `lang` (オプション): 結果の表のラベル（例えば「被説明変数」とか「観測数」とか）をどの**言語**で表示するかを指定する。`"ja"` (日本語)、`"en"` (英語)、`"zh"` (中国語) が選べる。デフォルトは `"ja"` なので、指定しなくても日本語で表示されることが多い。
*   `cov_type` (オプション): 標準誤差の計算方法、つまり**ロバスト標準誤差の種類**を指定する。デフォルトは `'HC1'` (小標本修正付きのWhiteの標準誤差) になっている。通常の標準誤差を使いたい場合は `'nonrobust'` を指定する。他にも `'HC0'`, `'HC2'`, `'HC3'` が選べるのは `statsmodels` と同じだ。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/14-48.png)

1.  `df_sample`: 実際に推定に使われたデータ（欠損値などが処理されたもの）。どんなデータで分析されたかを確認できる。
2.  `df_summary`: 推定に使われたデータの記述統計量の表（`regmonkey.summary()` と同じ形式）。データの基本的な性質をすぐに確認できる。
3.  `df_result`: **複数のモデルの推定結果が、比較しやすいように綺麗に整形された表**だ！これが我々のお目当て！ 

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/14-49.png)

*   **モデル1**：消費を所得だけで説明（基本的な単回帰）
*   **モデル2**：消費を所得と、**所得の二乗**で説明 (所得が増えるほど消費の増え方が鈍る、みたいな関係も見られるかも？)
*   **モデル3**：消費を所得と、**人口 (`人口`)** で説明する（重回帰分析の例）

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/14-50.png)

```python
# コード例14-11: まずは regmonkey をインストール
!pip install regmonkey
```

```python
# コード例14-12: 必要な関数をインポート
from regmonkey import regress, add_footer
```

```python
# コード例14-13
# 回帰モデルの設定：一度に複数のモデルを設定しても大丈夫
# --- 分析したいモデルの情報を、辞書のリストとして準備 ---

# モデル1：消費 ～ 所得
model1 = {"被説明変数": "消費", "説明変数": ["所得"]}  # 日本語のキーで指定

# モデル2：消費 ～ 所得 + 所得の2乗
model2 = {"被説明変数": "消費", "説明変数": ["所得", "所得**2"]}  # 所得の2乗項を追加

# モデル3：消費 ～ 所得 + 人口 (dfに'人口'列がある場合)
model3 = {"被説明変数": "消費", "説明変数": ["所得", "人口"]}

# すべての回帰モデルをリストとして、regressionに渡す
df_sample, df_summary, df_result = regress([model1, model2, model3], df)

# 推定系結果を表示
df_result
```

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/14-51.png)

*   **係数 (coef):** 最初の数行が各説明変数（および切片）が被説明変数に与える影響の大きさと方向を示す、推定された**回帰係数**の値だ。
*   **標準誤差 (std err):** 係数のすぐ下に括弧 `()` で囲まれて表示されているのが、係数の推定値の「ブレ具合」を示す**標準誤差**だ。
*   **有意性**：係数の横についている**アスタリスク `*`** は、その係数が統計的に有意であるかどうかを示している。通常、`***` なら1%水準、`**` なら5%水準、`*` なら10%水準で有意、という意味で、p値をいちいち見なくても、どの変数が効いていそうかパッと分かる便利な目印だ。
*   **決定係数と自由度修正済み決定係数**：モデルがデータの変動をどれだけ説明できているかを示す指標（0から1の値をとり、1に近いほど当てはまりが良い）。
*   **観測数**：回帰分析の推定に使われた**データの行数（サンプルサイズ）**だ。

## **14.6 今回の冒険のまとめ**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/14-59.png)

この章では、次のことを学びました。

* **回帰分析への招待**

  *   ある結果（**被説明変数**）が、他の要因（**説明変数**）によってどれくらい影響を受けるかを、**数式モデル**で明らかにする統計手法。
  *   **予測**や**要因分析**に活用できる。
  *   記述統計や可視化でデータの基本的な性質や関係性を事前に把握することが重要。

* **冒険の準備**

  *   **`statsmodels.formula.api` as `smf`**: 回帰分析を含む統計モデリングのための主要ライブラリ。`formula`形式でモデルを記述できる。
  *   データの読み込み（`pd.read_excel`など）と、`formula`で使いやすいように列名をシンプルにするなどの前処理。

* **`statsmodels` を使った回帰分析**

  *   **`formula`文字列の作成**: `'被説明変数 ~ 1 + 説明変数1 + 説明変数2'` のように、モデル構造を記述した文字列を準備（例: `my_formula` 変数に格納）。`1`は切片（定数項）を意味する。
  *   **モデルの設定**: `smf.ols(formula=my_formula, data=df)` で、OLS（最小二乗法）回帰モデルの準備（モデルオブジェクトの作成）。
  *   **モデルの推定**: `モデルオブジェクト.fit()` で、実際にデータを使って計算し、係数などを推定（結果オブジェクトの作成）。
  *   **推定結果の確認 (基本)**: `結果オブジェクト.params` で、推定された切片と各説明変数の回帰係数を取得。

  *   **仮説検定の考え方**: 「説明変数は影響を与えない（係数=0）」という帰無仮説を立て、データからその仮説がどれほど成り立ちにくいか（p値）を評価する。
  *   **`結果オブジェクト.summary()`**: 詳細な分析結果（係数`coef`、標準誤差`std err`、t値`t`、**p値`P>|t|`**、信頼区間`[0.025 0.975]`、決定係数`R-squared`、自由度修正済み決定係数`Adj. R-squared`など）を一覧表示。
  *   **p値と有意水準 ($\alpha$)**: p値が事前に設定した有意水準（例: 0.05）より小さければ、帰無仮説を棄却し、「統計的に有意な影響がある」と判断。
  *   **信頼区間**: 推定された係数の「ありそうな範囲」。0を跨いでいなければ、係数が0でないことを示唆。
  *   **決定係数 (R2) と自由度修正済み決定係数 (Adj. R2)**: モデルがデータの変動をどれだけ説明できているかを示す指標（0〜1）。Adj. R2は変数の数を考慮。
  *   **ロバスト標準誤差**: `fit(cov_type='HC1')` などで、誤差の不均一分散に対応した、より信頼性の高い標準誤差やp値を計算。

* **`regmonkey` ライブラリで効率化！**

  *   `from regmonkey import regress, add_footer`
  *   **`regress()` 関数**: 複数のモデル情報を辞書のリストで渡し、整形済みの比較可能な結果表を一括で取得。説明変数に `所得**2` (べき乗) や `log(所得)` (対数) といった変換も直接記述可能。
  *   結果のExcel保存 (`.to_excel()`) と `add_footer()` での注釈付けも容易に。
