#### **10.2.4.1 誕生時の初期化魔法：`__init__` メソッド (イニット メソッド)**

クラス定義の中に、`__init__` という特別な名前（アンダースコア`__`が前後に2つずつ付く）を持つ処理ブロックを書くと、それは**インスタンスが新しく作られるときに、自動的に1回だけ実行される特別な処理**として扱われるんだ。"initialize"（初期化する）の略だと考えるといいね。

この `__init__` の中で、そのインスタンスが持つべき属性（データ）の初期値を設定するのが一般的な使い方だ。

`__init__` の書き方は少し特別で、`def` キーワードを使って関数のように定義するんだけど、**最初の引数に『お決まりの言葉』として必ず `self` というものを書く**必要があるんだ。



```python
# コード例10-6
class Slime:
    # インスタンス生成時に自動で呼ばれる特別な初期化処理
    # 最初の self はお決まり。name と hp は外から受け取る情報。
    def __init__(self, name, hp):
        # ↓ ここでインスタンスに属性を持たせる
        self.my_name = name  # ① この個体の 'my_name' 属性に受け取った name を設定
        self.my_hp = hp  # ② この個体の 'my_hp' 属性に受け取った hp を設定
```

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/10-16.png)


> **Pythonにおける2つのタイプの命令（第3章　再掲）**
>
> 1. **関数 (Function):**
>    - `len()` や `print()`、`type()` のように、**単独で呼び出す**タイプ。
>    - 多くの場合、処理したいデータを括弧 `()` の中に**引数 (ひきすう)** として渡す。
>    - 例：`len(文字列)` → 文字列を `len` 関数に渡して長さを教えてもらう。
> 2. **メソッド (Method):**
>    - `.lower()` や `.strip()` のように、**データ（変数名など）の後ろに . (ドット) を付けて呼び出す**タイプ。
>    - そのデータ自身に「この操作をして！」と命令するイメージ。データに**所属している**専用の機能と言える。
>    - 例：`文字列.lower()` → 文字列自身に「小文字になって！」と命令する。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/10-17.png)


![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/10-18.png)


**`__init__` の中の `self.my_name = name`の意味**

- `self`: 「**今まさに作られている、このスライム自身**」
- `.my_name`: 「そのスライム自身が持つことになる **`my_name` という名前の属性（ステータス）**」
- `=` : 「に、以下の値を設定せよ」
- `name` (右側): 「`__init__` が呼び出されたときに**外から引数として渡された**値（例: `"スラリン"`）」

つまり、「**外から受け取った `name` の値を、自分自身 (`self`) の `my_name` 属性として設定しなさい**」という命令になるわけだ。`self.my_hp = hp` も同様に、「外から受け取った `hp` の値を、自分自身の `my_hp` 属性として設定しなさい」ということ。

`self` を使うことで、「これは他の誰でもない、**このインスタンス固有の**データですよ」ということを明確に示し、インスタンスにその属性を持たせているんだ。

#### **10.2.4.2 インスタンス化する際の引数の渡し方：`__init__` との連携**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/10-19.png)


```python
# コード例10-7
class Slime:
    # インスタンス生成時に自動で呼ばれる特別な初期化処理
    def __init__(self, name, hp):
        self.my_name = name
        self.my_hp = hp


# インスタンス化: Slime() の () に、__init__ の self を除いた引数 (name, hp) を渡す
# 1番目の実引数 "スラリン" が仮引数 name に
# 2番目の実引数 10 が仮引数 hp に渡される
slime_A = Slime("スラリン", 10)

# こちらも同様
# 1番目の実引数 "スラきち" が仮引数 name に
# 2番目の実引数 12 が仮引数 hp に渡される
slime_B = Slime("スラきち", 12)
```

**インスタンス化する際の引数の渡し方**

`Slime("スラリン", 10)` のようにクラス名の後ろの丸括弧 `()` に**`self` 以外の**引数を順番に渡すと、

- まず、Pythonは新しい `Slime` オブジェクト（インスタンス）をメモリ上に作成する。そして、その**新しく作られたオブジェクト自身**を、`__init__` メソッドの**最初の引数 `self` に自動的に渡す**んだ！
- さらに、`()`の中に書かれた `"スラリン"` や `10` といった**実引数**は、`__init__` メソッドの**`self` 以降の仮引数 (`name`, `hp`) に順番に渡される**。
- だから、プログラマ（我々）が `Slime("スラリン", 10)` と書くだけで、裏では `__init__(作られたばかりのスライム自身, "スラリン", 10)` が呼び出されている、というわけだね。

#### **10.2.5.1 メソッドの定義：クラスの中に `def` で、最初の引数は `self`！**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/10-25.png)


さあ、具体的に「体当たり攻撃 (`attack`)」と「ダメージを受ける (`damage`)」という2つの技（メソッド）を `Slime` クラスに追加してみよう。`__init__` も一緒に書いておくよ。

```python
# コード例10-10: メソッドを持つ Slime クラス
class Slime:
    def __init__(self, name, hp):
        self.my_name = name
        self.my_hp = hp

    # 体当たりするメソッド (技)
    def attack(self, target_name):  # 誰に攻撃するか(target_name)も受け取る
        print(
            f"{self.my_name}は {target_name}に 体当たりした！"
        )  # 自分の名前を使って攻撃メッセージ

    # ダメージを受けるメソッド (技)
    def damage(self, amount):
        original_hp = self.my_hp  # ダメージを受ける前のHPを一時的に保存
        self.my_hp -= amount  # 自分のHP(self.hp)を減らす
        
        # ダメージを受ける前と後のHPを表示
        print(f"{self.my_name}は {amount} のダメージを受けた！ (HP: {original_hp} → {self.my_hp})")

        # HPが0以下になったかどうかを判定
        if self.my_hp <= 0:
            print(f"{self.my_name}は 倒れてしまった...")
```

### **10.3.1 カプセル化 (Encapsulation)：大事なものを守るバリア**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/10-31.png)


![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/10-32.png)


では、HPを安全に管理する `SafeSlime` クラスを見てみよう。

```python
# コード例10-16: HPを安全に管理する SafeSlime クラス
class SafeSlime:
    def __init__(self, name, hp):
        # アンダースコア '_' を先頭につけることで、
        # 「これはクラス内部で使う属性なので、外部からは直接アクセスしないでね」
        # という開発者間の慣習的な目印（プライベート属性のような扱い）になる。
        # (※ Pythonでは完全に隠蔽されるわけではない)
        self._name = name
        self._hp = hp
        self._max_hp = hp  # 最大HPも属性として持っておく

    # HPを取得するためのメソッド (ゲッター Getter と呼ばれることもある)
    # 外部からはこのメソッドを通してHPを知るようにする
    def get_hp(self):
        return self._hp

    # ダメージを受けるメソッド (HPを安全に操作)
    def damage(self, amount):
        original_hp = self._hp  # ダメージを受ける前のHPを一時的に保存
        self._hp -= amount  # 自分のHP(self.hp)を減らす

        # ダメージを受ける前と後のHPを表示
        print(f"{self._name}は {amount} のダメージを受けた！ (HP: {original_hp} → {self._hp})")

        # HPがマイナスにならないようにチェック＆制御！
        if self._hp <= 0:
            print(f"{self._name} は倒れてしまった...")
            self._hp = 0  # HPを0で止める
        else:
            print(f"残りHP: {self._hp} / {self._max_hp}")

    # 回復するメソッド (HPを安全に操作)
    def heal(self, amount):
        # まず回復量を加算
        self._hp += amount
        # 最大HPを超えないようにチェック＆制御！
        if self._hp > self._max_hp:
            self._hp = self._max_hp  # 最大HPを超えないように制御
        print(f"{self._name} は {amount} 回復した！")
        print(f"現在HP: {self._hp} / {self._max_hp}")
```

さあ、この安全なスライムを使ってみよう。

```python
# コード例10-17: SafeSlimeクラスからインスタンスを作成。__init__ が呼ばれ、名前とHPが設定される
safe_slime = SafeSlime("スラぴょん", 20)
```

```python
# コード例10-18: メソッド経由でHPを取得
current_hp = safe_slime.get_hp()
print(f"メソッドで取得した現在のHP: {current_hp}")

# print(safe_slime._hp) # 直接アクセスも可能だが、非推奨
```

```python
# コード例10-20: healメソッドでHPを回復させる。HP0のスライムを10回復させてみる
safe_slime.heal(10)
print(f"回復後のHP (get_hpで確認): {safe_slime.get_hp()}")  # 10になっている
```

### **10.3.2 継承 (Inheritance)：親の能力を引き継いでパワーアップ！**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/10-37.png)


**ステップ1：親クラス（基本となるクラス）の準備**

まずは、継承の元となる親クラス `Slime` を用意しよう。（さっきのカプセル化を少し意識した `SafeSlime` を元にするよ）

```python
# コード例10-22: 親クラス：Slime (さっきの SafeSlime を少し変更)
class Slime:
    def __init__(self, name, hp, oridinary_slime=True):
        self._name = name
        self._hp = hp
        self._max_hp = hp

        # 通常のスライムなら、以下のメッセージを表示
        if oridinary_slime:
            print(f"通常のスライム '{self._name}' (HP:{self._hp}) が誕生！")

    def attack(self, target_name):
        print(f"{self._name} は {target_name} に 体当たり攻撃！")

    def damage(self, amount):
        original_hp = self._hp  # ダメージを受ける前のHPを一時的に保存
        self._hp -= amount  # 自分のHP(self.hp)を減らす

        # ダメージを受ける前と後のHPを表示
        print(f"{self._name}は {amount} のダメージを受けた！ (HP: {original_hp} → {self._hp})")

        # HPがマイナスにならないようにチェック＆制御！
        if self._hp <= 0:
            print(f"{self._name} は倒れてしまった...")
            self._hp = 0  # HPを0で止める
        else:
            print(f"残りHP: {self._hp} / {self._max_hp}")
```

**ステップ2：子クラス（新しいクラス）の定義と継承**

次に、`Slime` クラスを**継承**して `MetalSlime` クラスを定義する。継承するには、クラス名の後ろの**丸括弧 `()` の中に、親クラスの名前を書く**んだ。これが「このクラスは `Slime` を親としますよ」という宣言になる。

```python
# コード例10-23: 子クラス MetalSlime (Slimeクラスを継承)
class MetalSlime(Slime):  # ← カッコの中に親クラス名を書く！
    # メタルスライム独自の初期化処理を定義
    def __init__(self, name, hp, defense):
        # まず親クラスの __init__ を呼び出して、名前とHPを設定してもらう
        super().__init__(name, hp, oridinary_slime=False)  # super() で親クラスのメソッドを呼び出せる

        # --- 子クラス独自の属性を追加 ---
        self._defense = defense  # 子クラス独自の属性を追加
        print(f"メタルスライム '{name}' (守備力:{self._defense}) が現れた！")

    # 子クラス独自のメソッドを追加
    def flee(self):
        print(f"{self._name} は素早く逃げ出した！")

    # 親クラスの damage メソッドを上書き (オーバーライド)
    # メタルスライムは守備力でダメージを軽減する
    def damage(self, amount):
        actual_damage = max(1, amount - self._defense)  # 最低でも1ダメージは受ける
        print(f"メタルスライムの守備力({self._defense})でダメージを軽減！")
        # 親クラスの damage メソッドを呼び出して、実際のダメージ処理をしてもらう
        super().damage(actual_damage)
```

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/10-38.png)


**`super()` の使い方** 

*   `__init__` の中で `super().__init__(name, hp, oridinary_slime=False)` を呼び出しているのは、「メタルスライムとして生まれる前に、まず**親であるスライムとしての基本的な初期化（名前とHPの設定）を済ませてください。ただし、これは通常のスライムではないので、親クラス側の誕生メッセージはスキップしてください**」という指示を出している。これを省くと、メタルスライムが名前やHPを持てなくなり、さらに「通常のスライムが誕生」というメッセージも表示されてしまう。
*   `damage` メソッドの中で `super().damage(actual_damage)` を呼び出しているのは、「メタルスライム独自のダメージ軽減計算は終わったので、**残りのHPを減らしたり、倒れたかどうかの判定をする基本的な処理は、親クラスのスライムさん、お願いします**」と処理を委譲しているんだ。

このように `super()` を使うことで、親クラスの機能をうまく再利用しつつ、子クラス独自の変更や追加を効率よく行える。これが継承の強力な点の一つだよ。

### **10.3.3 多態性 (Polymorphism)：同じ呪文でも効果が変わる**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/10-42.png)


**ステップ1：様々なモンスタークラスを定義する**

まずは、`Slime` を親とする `MetalSlime` に加えて、全く別の親（例えば `Monster` という基本クラス）から派生するかもしれない `Dragon` クラスも作ってみよう。大事なのは、それぞれが `attack` という名前のメソッドを持つことだ。

```python
# コード例10-28

# 新しいモンスタークラス Dragon
class Dragon:
    def __init__(self, name, mp):
        self._name = name
        self._mp = mp
        print(f"ドラゴン '{self._name}' (MP:{self._mp}) が誕生！")

    def attack(self, target_name):
        # Slime と同じ名前の attack メソッドを定義
        print(f"《ドラゴン》 {self._name} は {target_name} に 炎を吐いた！ (ゴォー！)")


# 新しいモンスタークラス Golem
class Golem:
    def __init__(self, name, power):
        self._name = name
        self._power = power
        print(f"ゴーレム '{self._name}' (パワー:{self._power}) が誕生！")

    def attack(self, target_name):
        # これも同じ名前の attack メソッド
        print(
            f"《ゴーレム》 {self._name} は {target_name} に 岩石を投げつけた！ (ドシーン！)"
        )
```

**ステップ2：異なる種類のモンスターをリストにまとめる**

次に、これらの異なる種類のモンスター（オブジェクト）を、一つのリストに入れてみよう。

```python
# コード例10-29: いろんなモンスターをリストに入れる

# それぞれのクラスからインスタンスを作成
slime1 = Slime("スラりん", 10)
metal1 = MetalSlime("メタぞう", 5, 10)
dragon1 = Dragon("ドラキチ", 50)
golem1 = Golem("ゴムタン", 80)

# 作成したインスタンスを一つのリストにまとめる
monsters = [slime1, metal1, dragon1, golem1]
```

**ステップ3：ループで全員に同じ命令を出す！**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/10-44.png)


```python
# コード例10-30: ループで各モンスターに攻撃させる
for monster in monsters:
    # monster 変数には Slime, MetalSlime, Dragon, Golem が順番に入る
    # しかし、呼び出す側はどの種類のモンスターか気にする必要はない！ただ attack() メソッドを呼び出すだけ！
    monster.attack("壁")  # ★ここが多態性！ 同じ呼び出し方で違う動作！
```

この章では、次のことを学びました。

*   **オブジェクト:**
    *   データ（**属性**）と機能（**メソッド**）がセットになった「モノ」。
    *   Pythonのすべてのデータ型（`str`, `list` など）はオブジェクト。
    *   `オブジェクト.メソッド()` で、そのオブジェクト専用の技を使う。
*   **クラス:**
    *   オブジェクトの**設計図**または**テンプレート** (`class` で定義)。
    *   クラスからオブジェクト（**インスタンス**）を作ることを**インスタンス化** (`クラス名()`)。
    *   `__init__(self, ...)`: インスタンス生成時に呼ばれ、属性を初期化する特別なメソッド。`self` はインスタンス自身。
    *   クラス内で定義した関数が**メソッド**になる（第一引数は `self`）。
*   **オブジェクト指向:**
    *   プログラムを独立した「オブジェクト」の集まりとして捉える**設計思想**。
    *   **三大要素**: カプセル化（隠蔽）、継承（引き継ぎ）、多態性（多様な振る舞い）。
