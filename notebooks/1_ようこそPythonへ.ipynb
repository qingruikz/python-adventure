Python初心者が print() を使う際によくあるエラーをまとめました：

| よくあるミス | エラーメッセージ | 原因と対策 |
| ---- | ---- | ---- |
| `primt("こんにちは")` | NameError: name 'primt' is not defined | `print` の スペルミス。mではなく、正しくは `print`。 |
| `print（"こんにちは"）` | SyntaxError: invalid character '（' (U+FF08)     | 全角の `（）` を使ったため。英文字や括弧は必ず**半角文字**で入力しましょう。 |
| `print("こんにちは)` | SyntaxError: unterminated string literal (detected at line 1) | 最後のダブルクォーテーションが抜けている。文字は `"` で始まり、必ず `"` で閉じましょう。 |
| `print("こんにちは"` | SyntaxError: incomplete input | 最後の `)` が抜けている。`(` で始まった括弧は、必ず `)` で閉じましょう。 |

焦らずにエラーメッセージを読み、上記の点を確認してみてください。

## **コラム Pythonの由来**

Pythonの直訳はニシキヘビだが、ヘビとは直接関係ない。開発者であるオランダ人のグイド・ヴァンロッサム氏がイギリスのコメディ番組「モンティ・パイソン (Monty Python)」の大ファンだったことから名付けられたんだそうだ。グイドが1989年のクリスマス休暇に余暇を利用し、Pythonの開発を始めた。Pythonの簡潔さと効率性を表す名言に「Life is short, you need Python. (人生は短い、だからPythonを使おう。)」がある。

### **1.4.1 開発の流れ：宝の地図を描くように**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/1-18.png)

**開発の流れ**
1.  **計画 (Plan)：** 何を作りたいのか、どんな機能が必要か、どういう手順で実現するかを考える。（宝の地図の目的地とルートを決める）
2.  **コーディング (Code)：** 計画に基づいて、実際にプログラムのコードを書く。（地図に従って道を進む）
3.  **テスト (Test)：** 書いたコードが思った通りに動くか、色々な状況で試してみる。（道が合っているか、危険がないか確認する）
4.  **デバッグ (Debug)：** テストで見つかったエラーや問題点（バグと呼ぶ）を修正する。（道を塞ぐ障害物を取り除く）
5.  **完成 (Complete)：** テストとデバッグを繰り返し、プログラムが完成する。（宝箱にたどり着く！）

### **1.4.3 プログラムの書き方：魔法の呪文のルール**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/1-23.png)

*   **半角英数字：** プログラムの命令（`print`など）や名前（変数名など、後で学ぶよ）は、基本的に**半角の**英数字と記号で書く必要がある。全角文字を使うと、エラーの原因になることが多い。
*   **大文字と小文字の区別：** Pythonは、大文字と小文字を厳密に区別する。例えば、`print` と `Print` は違うものとして扱われる。さっき拓海くんが `NameError` を出したようにね。
*   **インデント（字下げ）：** 行の先頭のスペース（空白）のことをインデントと呼ぶ。Pythonでは、このインデントが非常に重要で、**プログラムの構造**を示す役割を持っているんだ。これは今後の章で詳しく学んでいくが、今は「行の先頭にむやみにスペースを入れない」ことを覚えておこう。
*   **正確さ：** コンピューターは曖昧な指示を理解できない。括弧 `()` の閉じ忘れ、カンマ `,` の打ち間違い、スペルミスなど、些細な間違いでもエラーになる。注意深く書くことが大切だ。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/1-24.png)

### **1.4.4 コメント：未来の自分や仲間へのメッセージ**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/1-25.png)

Pythonでは、シャープ記号 `#` から行末までがコメントとして扱われます（緑色の文字で表示される）。

Pythonがプログラムを実行する際、これらのコメントは完全に無視される。

```python
# コード例1-4: これはコメントです。プログラムとしては無視されます。
print("この行は実行されます")  # ここから右もコメントです

# print("この行はコメントアウトされているので実行されません")
```

**コメントの役割：**

*   **コードの説明：** そのコードが何をしているのか、なぜそう書いたのかを説明する。
*   **一時的にコードを無効化（コメントアウト）：** コードを削除せずに、一時的に実行されないようにする。
*   **備忘録：** 後で修正する箇所や、注意点をメモしておく。
*   **他の人へのメッセージ：** チームで開発する際に、他のメンバーにコードの意図を伝える。
