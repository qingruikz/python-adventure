## **7.1 辞書の特徴：キーと値のペアで管理**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/7-2.png)


### **7.1.1 辞書（ディクショナリ）とは？**

*   **キー (Key):** 値を探すための「見出し」や「名前」となるもの。**辞書の中で重複してはいけない**。通常、文字列や数値、タプルなど**変更不可能な (immutable)** データ型を使う。
*   **値 (Value):** キーに対応付けられたデータ。こちらは数値、文字列、リスト、別の辞書など、**どんなデータ型でもOK**。
*   **ペア (Item):** 1つのキーと値の組。
*   **順番:** Python 3.7以降では、辞書に追加された要素の順番がある程度保持されるようになったが、基本的にはリストやタプルのように「インデックス番号による順番」を前提とした操作は行わない。**キーを使って値にアクセスする**のが基本。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/7-3.png)


### **7.1.2 辞書の作成**

辞書は、波括弧 `{}` を使い、`キー: 値` というペアをカンマ `,` で区切って作成する。

```python
# コード例7-1: 空の辞書を作成
empty_dict = {}
print(empty_dict)
```

```python
# コード例7-2: 辞書のタイプ
type(empty_dict)
```

```python
# コード例7-3: 生徒の情報を格納した辞書
student = {"name": "佐藤 拓海", "age": 20, "major": "経済学"}
print(student)
```

```python
# コード例7-4: 商品価格の辞書 (キーが文字列、値が数値)
prices = {"リンゴ": 120, "バナナ": 80, "オレンジ": 150}
print(prices)
```

```python
# コード例7-5: キーが数値の辞書 (あまり一般的ではないが、可能)
number_names = {1: "One", 2: "Two", 3: "Three"}
print(number_names)
```

### **7.1.3 辞書の要素数（ペアの数） (`len` 関数)**

辞書にいくつの「キーと値のペア」が含まれているかは、リストや文字列と同じく `len()` 関数で調べられる。

```python
# コード例7-6
student = {"name": "佐藤 拓海", "age": 20, "major": "経済学"}
prices = {"リンゴ": 120, "バナナ": 80, "オレンジ": 150}
empty_dict = {}

print(f"student の要素数 (ペア数): {len(student)}")
print(f"prices の要素数: {len(prices)}")
print(f"empty_dict の要素数: {len(empty_dict)}")
```

### **7.1.4 キーによる値の取得：名前で中身を取り出す！**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/7-5.png)


#### **方法1：角括弧 `[]` を使う**


`辞書変数[キー]` のように、角括弧の中にキーを指定する。

```python
# コード例7-7: キー "name" に対応する値を取得
student = {"name": "佐藤 拓海", "age": 20, "major": "経済学"}
print(student["name"])
```

```python
# コード例7-8: キー "age" に対応する値を取得
print(student["age"])
```

```python
# コード例7-9: わざとエラーを起こしてみる：存在しないキーを指定するとどうなる？ -> エラー！
student["hobby"]  # "hobby" というキーは存在しない
```

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/7-6.png)


#### **方法2：`.get()` メソッドを使う**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/7-7.png)


さらに、`.get(キー, デフォルト値)` のように第二引数を指定すると、**キーが存在しない場合に `None` の代わりに指定したデフォルト値を返す**ことができる。

```python
# コード例7-10: 存在するキー "name" を取得
student = {"name": "佐藤 拓海", "age": 20, "major": "経済学"}

name = student.get("name")
print(name)
```

```python
# コード例7-11: 存在しないキー "hobby" を取得 -> エラーにならない！
hobby = student.get("hobby")
print(hobby)
```

```python
# コード例7-12: 存在しないキー "hobby" を取得し、デフォルト値を指定
hobby_default = student.get("hobby", "不明")  # キーがなければ "不明" を返す
print(hobby_default)
```

## **7.2 辞書の基本操作：棚の整理と中身の入れ替え**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/7-9.png)


### **7.2.1 要素の変更・追加**

辞書の要素（値）を変更したり、新しいキーと値のペアを追加したりするのは、非常に簡単だ。角括弧 `[]` を使ってキーを指定し、代入演算子 `=` で値を設定する。

*   **キーが既に存在する場合:** そのキーに対応する**値が上書き**される。
*   **キーが存在しない場合:** そのキーと値の**新しいペアが辞書に追加**される。

```python
# コード例7-13: 既存のキーの値を変更 (age を 21 に)
student = {"name": "佐藤 拓海", "age": 20}
print(f"変更前: {student}")

student["age"] = 21
print(f"年齢変更後: {student}")
```

```python
# コード例7-14: 新しいキーと値のペアを追加 (major を追加)
student["major"] = "経済学"
print(f"専攻追加後: {student}")
```

```python
# コード例7-15: さらに新しいペアを追加 (hobby を追加)
student["hobby"] = "ゲーム"
print(f"趣味追加後: {student}")
```

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/7-10.png)


### **7.2.2 要素の削除 (`del`, `.pop()`, `.clear()`)**

辞書から不要になったキーと値のペアを削除する方法もいくつかある。

*   `del 辞書[キー]`: 指定した**キー**とその値のペアを削除する。存在しないキーを指定すると `KeyError`。
*   `辞書.pop(キー)`: 指定した**キー**のペアを削除し、**削除した値**を返す。存在しないキーを指定すると `KeyError` になるが、`pop(キー, デフォルト値)` とすれば、キーがない場合にデフォルト値を返しエラーにならない。
*   `辞書.clear()`: 辞書の**全てのキーと値のペア**を削除し、空の辞書にする。

```python
# コード例7-16: del でキー "major" を削除
student = {"name": "佐藤 拓海", "age": 21, "major": "経済学", "hobby": "ゲーム"}
print(f"削除前の辞書: {student}")

del student["major"]
print(f"del student['major'] 後: {student}")
```

```python
# コード例7-17: .pop() でキー "hobby" の値を返しつつ削除
removed_hobby = student.pop("hobby")
print(f".pop('hobby') 後: {student}")
print(f"removed_hobby: {removed_hobby}")
```

```python
# コード例7-18: .pop() で存在しないキーをデフォルト値付きで試す
removed_club = student.pop("club", "未所属")  # "club"キーはないのでデフォルト値が返る
print(f"removed_club: {removed_club}")
print(f"student: {student}")
```

```python
# コード例7-19: .clear() で全要素を削除
student.clear()
print(student)
```

## **7.3 辞書の応用：繰り返し処理と便利な使い方**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/7-12.png)


### **7.3.1 キーの存在確認 (`in`)**

特定の**キー**が辞書に含まれているかどうかを調べるには、リストと同じように `in` 演算子が使える。

```python
# コード例7-20: キーが存在するか
student = {"name": "佐藤 拓海", "age": 20}

print(f"'name' キーは存在するか？ {'name' in student}")
print(f"'major' キーは存在するか？ {'major' in student}")
```

```python
# コード例7-21: not in も使える
print(f"'hobby' キーは存在しないか？ {'hobby' not in student}")
```

**注意:** `in` 演算子はデフォルトでは**キーの存在**をチェックする。値が含まれているかを直接チェックするわけではない点に注意しよう。

### **7.3.2 `for`文による繰り返しと辞書**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/7-13.png)


#### **パターン1：キーを順番に取り出す**

`for`文で辞書をそのまま使うと、デフォルトでは**キー**が順番に取り出される。

```python
# コード例7-22: キーの繰り返し
prices = {"リンゴ": 120, "バナナ": 80, "オレンジ": 150}

for key in prices:
    # キーを取り出す
    print(f"キー: {key}")
    # キーを使って値を取得できる
    value = prices[key]
    print(f"  -> 値: {value}")
```

#### **パターン2：`.keys()` メソッドでキーを取り出す**

辞書には、キーだけ、値だけ、あるいはキーと値のペアを順番に取り出すための専用のメソッドがある。`.keys()` メソッドを使うと、明示的にキーの一覧を取り出して繰り返せる（結果はパターン1と同じ）。

```python
# コード例7-23: .keys() でキーを取り出す
for key in prices.keys():
    print(f"キー: {key}")
```

#### **パターン3：`.values()` メソッドで値を取り出す**

`.values()` メソッドを使うと、**値**だけを順番に取り出して繰り返せる。

```python
# コード例7-24: .values() で値を取り出す
total_price = 0
for value in prices.values():
    print(f"値段: {value}")
    total_price += value
print(f"合計金額: {total_price}")
```

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/7-14.png)


#### **パターン4：`.items()` メソッドでキーと値のペアを取り出す**

`.items()` メソッドを使うと、**キーと値のペア（タプルとして）** を順番に取り出して繰り返せる。`for`文で2つの変数（キー用、値用）を使って受け取るのが一般的だ。これが一番よく使われるかもしれない。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/7-15a.png)


```python
# コード例7-25: .items() でキーと値のペアを取り出す
for key, value in prices.items():  # キーをkey変数、値をvalue変数で受け取る
    print(f"商品名(キー): {key}, 価格(値): {value}円")
```

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/7-15b.png)


### **7.3.3 辞書内包表記 (Dictionary Comprehension) (発展)**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/7-16.png)


`{新しいキー: 新しい値 for 変数 in イテラブル}` のような形が基本だ。

```python
# コード例7-26: 通常のループで0から4までの数をキー、その2乗を値とする辞書を作成
squares_dict_loop = {}
for i in range(5):
    squares_dict_loop[i] = i * i

print(f"\n通常ループで作った辞書: {squares_dict_loop}")
```

```python
# コード例7-27: 辞書内包表記
squares_dict_comp = {i: i * i for i in range(5)}
print(f"内包表記で作った辞書: {squares_dict_comp}")
```

### **7.3.4 辞書とリストの比較**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/7-18.png)


| 特徴                 | リスト (`list`)                              | 辞書 (`dict`)                                           |
| :------------------- | :------------------------------------------- | :------------------------------------------------------ |
| **データの格納**     | 複数の要素を**順番に**格納                   | キーと値の**ペア**を格納                                |
| **要素へのアクセス** | **インデックス番号** (`[0]`, `[1]`, ...)     | **キー** (`[キー]`, `.get(キー)`)                       |
| **順序**             | あり（追加された順）                         | 基本的になし（Python 3.7+では追加順保持）               |
| **変更可能性**       | 変更可能 (mutable)                           | 変更可能 (mutable)                                      |
| **キーの制約**       | なし                                         | キーは**一意**で**変更不可能(immutable)**である必要あり |
| **主な用途**         | 順番が重要なデータの集まり、大量データの格納 | 名前やIDでデータを管理、設定情報、項目ごとの値          |

**どんな時にどちらを使うか？**

*   データの**順番**が重要だったり、単に複数のものを**まとめて**扱いたい場合は **リスト**。
*   各データに**名前（キー）**を付けて、その名前で**効率よく**情報を取り出したい場合は **辞書**。

## **7.4 今回の冒険のまとめ**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/7-19.png)


この章では、次のことを学びました。

*   **辞書 (`dict`) の特徴:**
    *   **キー**と**値**のペアでデータを格納するデータ型。
    *   波括弧 `{}` と `キー: 値` で作成 (`{}` は空辞書)。
    *   要素（ペア）の順番は基本的には保証されない（キーでアクセスするため）。
    *   キーは一意で変更不可能 (immutable) である必要あり。値は何でもOK。
    *   `len(辞書)` でペア数を取得。
    *   値の取得: `辞書[キー]` (キーがないと`KeyError`) または `辞書.get(キー, デフォルト値)` (エラーにならない)。
*   **辞書の基本操作 (変更可能):**
    *   変更・追加: `辞書[キー] = 値` (キーがあれば上書き、なければ新規追加)。
    *   削除: `del 辞書[キー]`, `辞書.pop(キー, デフォルト値)` (値を返す), `辞書.clear()` (全削除)。
*   **辞書の応用:**
    *   キーの存在確認: `キー in 辞書`。
    *   `for`文での繰り返し:
        *   `for key in 辞書:` (キーが取り出される)
        *   `for key in 辞書.keys():` (キーが取り出される)
        *   `for value in 辞書.values():` (値が取り出される)
        *   `for key, value in 辞書.items():` (キーと値のペア(タプル)が取り出される)
    *   辞書内包表記: `{キー式: 値式 for ...}` (発展)。
    *   リストとの比較: 順番 vs キー、アクセス方法、用途の違い。
