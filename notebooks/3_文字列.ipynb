### **3.1.3 文字へのアクセス（インデックス）**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/3-5.png)


**重要な注意点：** Pythonのインデックスは、**0から始まる！** つまり、最初の文字のインデックスは `0`、2番目の文字は `1`、3番目の文字は `2`…となるんだ。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/3-6.png)


変数名の後ろに角括弧 `[]` をつけ、その中にインデックス番号を指定することで、その位置の文字を取り出せる。

```python
# コード例3-7: 最初の文字

word = "Python"

# 最初の文字 (インデックス 0)
char0 = word[0]
print("最初の文字は：")
print(char0)
```

```python
# コード例3-8: 3番目の文字 (インデックス 2)
char2 = word[2]
print("3番目の文字は：")
print(char2)
```

```python
# コード例3-9: 最後の文字は？ len()を使うと便利

# 長さが6なら最後のインデックスは5
last_index = len(word) - 1  
char_last = word[last_index]
print("最後の文字は：")
print(char_last)
```

```python
# コード例3-10: 負のインデックスも使える！ -1 は最後の文字、-2 は最後から2番目
char_second_last = word[-2]
print("最後から2番目の文字：")
print(char_second_last)
```

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/3-7.png)


```python
# コード例3-11: わざと範囲外のインデックスを指定してみる
char6 = word[6]
print("6番目の文字：")
print(char6)
```

### **3.1.4 部分的な取り出し（スライス）**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/3-9.png)


スライスは、角括弧 `[]` の中にコロン `:` を使って `[開始インデックス:終了インデックス]` のように指定する。

**重要な注意点：**

*   開始インデックスの文字は**含まれる**。
*   終了インデックスの文字は**含まれない**（その手前まで）。
*   開始インデックスを省略すると、文字列の最初からになる。
*   終了インデックスを省略すると、文字列の最後までになる。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/3-10.png)


スライスの例をいくつか見てみよう。

```python
# コード例3-12 
long_word = "Programming"
len(long_word)
```

`len(long_word)` は `11` なので、インデックスは `0` から `10`。

```python
# コード例3-13: 2文字目(インデックス1)から4文字目(インデックス3)までを取り出す：インデックス 1, 2, 3 が対象。終了インデックスは4を指定
slice1 = long_word[1:4]
print(slice1)
```

```python
# コード例3-14: 最初から5文字目(インデックス4)までを取り出す：インデックス 0, 1, 2, 3, 4 が対象。開始を省略したら、最初から。終了インデックスは5を指定。
slice2 = long_word[:5]
print(slice2)
```

```python
# コード例3-15: 8文字目(インデックス7)から最後までを取り出す：インデックス 7, 8, 9, 10 が対象。終了を省略したら、最後まで。
slice3 = long_word[7:]
print(slice3)
```

```python
# コード例3-16: 全体をコピー（あまり使わないが）
slice_all = long_word[:]
print(slice_all)
```

### **3.1.5 文字列の検索**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/3-12.png)


そんな時に役立つのが `in` 演算子と `not in` 演算子だ。これらは、文字列の中に指定した部分文字列が存在するかどうかを調べて、`True` (存在する) または `False` (存在しない) の真偽値を返してくれる。

```python
# コード例3-17: 文字列の検索 

sentence = "Python is fun and powerful!"

# "fun" という単語が含まれているか？
contains_fun = "fun" in sentence
print(contains_fun)
```

### **3.1.6 変数と文字列の混乱：箱の名前と、ただの言葉**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/3-14.png)


例を見てみよう。生徒の名前を変数 `student_name` に入れて、挨拶を表示したいとする。

```python
# コード例3-19: 変数の中身を利用

# 生徒の名前を変数に代入
student_name = "高橋さん"

# 変数を使って挨拶を表示（正しい使い方）
print("こんにちは、")
print(student_name)
```

```python
# コード例3-20: 変数名自体を文字列として表示（よくある間違い）
print("こんにちは、")
print("student_name")
```

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/3-15.png)


**変数と文字列の違い**

- **クォート無し** (`student_name`): これは「`student_name` という名前の**変数**（箱）の中身を参照してね」という命令になる。だから、中に入っている `"高橋さん"` という値が使われる。
- **クォート有り** (`'student_name'` や `"student_name"`): これは「student_name という**文字列リテラル**（文字の並び）そのものを扱ってね」という命令になる。変数とは全く関係なく、「`s`, `t`, `u`, ...」という文字が表示される。

### **3.2.2 エスケープシーケンス：改行や特殊記号を埋め込む**

### **3.3.2 f-string：現代的な文字列フォーマット術！**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/3-29.png)


使い方はシンプルだ。

1. 文字列の開始クォート（`'` または `"`）の**直前に `f` を書く**。
2. 文字列の中で、変数や式を埋め込みたい箇所を**波括弧 `{}` で囲む**。

これだけだ！ さっきの例を f-string で書き直してみよう。

```python
# コード例3-38: f-string
item_name = "リンゴ"
price = 120
quantity = 5

# f-string を使う場合
message_f = f"{item_name}を{quantity}個買いました。合計金額は{price * quantity}円です。"
print(message_f)
```

### **3.3.3 大文字・小文字の変換**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/3-31.png)


**大文字・小文字の変換用命令**

*   `.lower()`: 文字列を**すべて小文字**に変換した新しい文字列を返す。
*   `.upper()`: 文字列を**すべて大文字**に変換した新しい文字列を返す。
*   `.capitalize()`: 文字列の**先頭の文字だけを大文字**にし、残りを小文字にした新しい文字列を返す。

```python
# コード例3-39
mixed_case = "PyThOn Is FuN"
print(f"元の文字列: '{mixed_case}'")
```

```python
# コード例3-40: 全て小文字に変換。mixed_case に .lower() を適用。
lower_case = mixed_case.lower()
print(lower_case)
```

```python
# コード例3-41: 全て大文字に変換。mixed_case に .upper() を適用。
upper_case = mixed_case.upper()
print(upper_case)
```

```python
# コード例3-42: 先頭だけ大文字、残りは小文字。mixed_case に .capitalize() を適用。
capitalized = mixed_case.capitalize()
print(capitalized)
```

```python
# コード例3-43: 元の文字列は変わらないことに注意！
print(mixed_case)
```

### **3.3.4 2つのタイプの命令：関数とメソッド**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/3-33.png)


**Pythonにおける2つのタイプの命令**

1. **関数 (Function):**
   - `len()` や `print()`、`type()` のように、**単独で呼び出す**タイプ。
   - 多くの場合、処理したいデータを括弧 `()` の中に**引数 (ひきすう)** として渡す。
   - 例：`len(文字列)` → 文字列を `len` 関数に渡して長さを教えてもらう。
2. **メソッド (Method):**
   - `.lower()` や `.strip()` のように、**データ（変数名など）の後ろに . (ドット) を付けて呼び出す**タイプ。
   - そのデータ自身に「この操作をして！」と命令するイメージ。データに**所属している**専用の機能と言える。
   - 例：`文字列.lower()` → 文字列自身に「小文字になって！」と命令する。

この章では、次のことを学びました。

*   **文字列の基本:**
    *   文字列は文字の**順序付き**の並び (`str`)。
    *   `len(文字列)` で長さを取得。
    *   `文字列[インデックス]` で特定位置の文字を取得 (**0始まり**)。
    *   `文字列[開始:終了]` で部分文字列を切り出す (**終了は含まない**)。
    *   `in` / `not in` で部分文字列が含まれるか/含まれないかをチェック (`True`/`False`)。
    *   数字に見える文字列 (`"123"`) と数値 (`123`) は別物。
*   **特殊な文字列:**
    *   空文字列 (`""`, `''`) は長さ `0` の文字列。

    *   `r"..."` (raw文字列) で `\` をそのまま扱える。
*   **文字列の加工:**
    *   `+` で文字列同士を連結。
    *   f-string (`f"..."`) で変数や式を埋め込み。
    *   **メソッド**: データに `.` を付けて呼び出す命令。
        *   `.lower()`, `.upper()`, `.capitalize()`: 大文字/小文字変換。
        *   メソッドは通常、**元の文字列を変更せず、新しい文字列を返す**。
    *   **関数とメソッドの違い**: `len(データ)` vs `データ.メソッド()`。
