### **12.2.4 グラフを一括でお洒落に！ `seaborn` のスタイル設定 (`sns.set_theme`)**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/12-19.png)


```python
# コード例12-5: seaborn の描画スタイルを設定
sns.set_theme(
    style="whitegrid",  # グラフの背景スタイル。'darkgrid'(背景暗,グリッド有), 'white'(背景白,グリッド無), 'ticks'(軸に目盛り) など。
    font="IPAexGothic",  # ★重要: japanize_matplotlib をインポートした後、ここで日本語フォントを指定！
    font_scale=1.3,  # 文字や数字の基本サイズを1.3倍にする（見やすくするため）。
    rc={  # rcParams の略。 matplotlib のより細かい設定項目を辞書形式で指定できる。
        "figure.figsize": (
            12,
            7,
        ),  # グラフのデフォルトサイズを横12インチ、縦7インチに。
        "axes.grid": True,  # グリッド線を表示する (style='whitegrid'ならデフォルトでTrue)。
        "grid.linestyle": "--",  # グリッド線を破線にする。 (デフォルトは実線)
        "axes.linewidth": 1.2,  # 座標軸の線の太さ。
        "axes.edgecolor": "gray",  # 座標軸の枠線の色を濃い灰色に。
    },
)

print("seabornのスタイル設定完了！これでグラフが綺麗になるはず！")
```

### **12.5.1 データの準備：特定の地域の時系列データを抽出**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/12-32.png)


**`.query()` メソッドの基本的な使い方**：`DataFrame変数.query("条件式を文字列で書く")`

*   `'条件式を文字列で書く'` の部分には、列名と値を比較する式を書く。
    *   例: `"地域 == '北海道 札幌市'"` (地域列が `"北海道 札幌市"` と完全に一致する行)
    *   例: `"年度 >= 2000"` (年度列が `2000` 以上の行)
    *   複数の条件は `and` や `or` で繋げられる。
*   **重要注意点:**
    *   文字列の値と比較する場合は、その値をさらにクォーテーション（シングル `'` またはダブル `"`）で囲む必要がある。条件式全体が文字列なので、内側のクォートは外側と違う種類を使うのが一般的だ (例: `query("地域 == '北海道 札幌市'")` や `query('地域 == "北海道 札幌市"')`)。
    *   列名に**スペース、特殊文字（括弧`()`やハイフン`-`など）、あるいは列名が数字で始まる場合**は、その列名を**バッククォート `` ` ``** で囲む必要がある。
        *   例: ``query("`GDP(2025年度)` < 100")``
        *   例: ``query("`列名に スペース` == "値"")``
        *   例: ``query("`15歳未満人口割合` > 10")``

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/12-33.png)


```python
# コード例12-14

# 特定の地域のデータを抽出
df_line = df.query("地域 == '北海道 札幌市' or 地域 == '北海道 夕張市'")

# 抽出結果をすべて表示（6年×2団体分しかない）
df_line
```

### **12.5.3 データの並び替え：綺麗な折れ線を描くための下準備**

**`.sort_values()` の使い方 (復習):** `DataFrame変数.sort_values(by=['基準列1', '基準列2', ...], ascending=True/False)`

1. 「`地域`」列が、我々が指定するリスト（`['北海道 札幌市', '北海道 夕張市', ...]`）の**いずれか**であること。これは `in` 演算子が使えるね。

### **12.7.1 ヒストグラムの仕組み：「ビン」でデータを仕分ける**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/12-51.png)


例えば、「15歳未満人口割合」のデータが5%から20%まであるとして、これを「5%以上7%未満」「7%以上9%未満」…のように2%ずつの区間（ビン）に分ける。そして、各ビンにデータがいくつ入っているかを数え上げ、その数を棒の高さとしてグラフにする。これがヒストグラムの基本的な仕組みだよ。

`seaborn` ライブラリでヒストグラムを描く基本的な呪文は `sns.histplot()` だ。

**`sns.histplot()` の主な引数:**

*   `data=DataFrame名`: グラフの元になるデータが入ったDataFrameを指定する。
*   `x="列名"`: 横軸（X軸）にしたい列の名前を文字列で指定する。通常ここに分布を見たい列（例：`"15歳未満人口割合"`）を指定する。

    *   `'probability'`: 各ビンの度数を総度数で割った**確率（相対度数）**。全ての棒の高さを足すと1になる。
    *   `'percent'`: 確率を100倍した**パーセント表示（相対度数%）**。全ての棒の高さを足すと100になる。

**`.strftime(書式文字列)` メソッドとは？**：`datetime`オブジェクトが持つ、日時情報を指定した「書式文字列」に従って文字列に変換するためのメソッドだよ。

*   `%Y`: 4桁の年 (例: 2023)
*   `%m`: 2桁の月 (例: 01, 10)
*   `%d`: 2桁の日 (例: 05, 27)
*   `%H`: 24時間表記の時 (例: 09, 15)
*   `%M`: 2桁の分 (例: 00, 30)

```python
# 日本のタイムゾーン（JST）を指定して、現在の日本の日時を取得し、'202510271530'（年月日時分）形式の文字列に変換
NOW = datetime.now(ZoneInfo("Asia/Tokyo")).strftime('%Y%m%d%H%M')
print(f"現在の日本時刻の文字列: {NOW}")
```

### **12.9.2 グラフの保存 (`plt.savefig`)**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/12-74.png)


**`plt.savefig()`の基本的な使い方と主な引数:** `plt.savefig('保存ファイルパス', dpi=解像度, bbox_inches='tight')`

*   **`'保存ファイルパス'` (最初の引数):** これが**必須**の引数だ。**どのフォルダに、どんな名前で、どんな種類のファイルとして保存するか**を文字列で指定する。
    *   ファイル名には、`.png` (背景透過も可能な高品質画像)、`.jpg` (写真などでよく使われる圧縮画像)、`.pdf` (印刷に適したベクター画像)、`.svg` (拡大縮小しても劣化しないベクター画像) といった**拡張子**を付けることで、保存形式を指定できる。通常は `.png` が扱いやすくておすすめだ。
    *   我々がさっき作った `出力ファイル/現在時刻` フォルダのパスと、グラフの内容が分かるファイル名を組み合わせて指定しよう。
*   `dpi=数値`: (オプション) **解像度 (dots per inch)** を指定する。数値が大きいほど、より高画質で詳細な画像になるが、ファイルサイズも大きくなる。ウェブで使うなら `100`～`150`、印刷物やポスターに使うなら `300` 以上を指定することが多い。デフォルトは通常 `100` 程度だ。
*   `bbox_inches='tight'`: (オプション) これも**非常に便利なオプション**だ！ これを指定すると、グラフのタイトルや軸ラベル、凡例などが画像の外にはみ出さないように、**周囲の余白を自動的にトリミングして、グラフがぴったり収まるように**保存してくれる。指定しないと、余白が大きすぎたり、逆にラベルが切れてしまったりすることがある。
