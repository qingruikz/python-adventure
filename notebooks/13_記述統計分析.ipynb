### **13.3.1 最初のステップ：`.describe()` でデータの健康診断！**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/13-13.png)


```python
# コード例13-5: 基本統計量を計算
df.describe()
```

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/13-14.png)


*   `count`: **データの個数**（欠損値 `NaN` は数えない）。
*   `mean`: **平均値**。お馴染みの、全データを足して個数で割った値だね。
*   `std`: **標準偏差 (Standard Deviation)**。データが平均値からどれくらい散らばっているかを示す代表的な指標。値が大きいほど、データが広範囲にばらついていることを意味する。
*   `min`: **最小値**。その列で一番小さい値。
*   `25%` (第1四分位数): データを小さい順に並べたとき、全体の**下から25%**の位置に来る値。
*   `50%` (第2四分位数): データを小さい順に並べたとき、全体の**ちょうど真ん中 (50%)** の位置に来る値。これは**中央値 (Median)** とも呼ばれる。平均値は極端に大きい値や小さい値（外れ値）に影響されやすいが、中央値はそういった影響を受けにくいという特徴がある。
*   `75%` (第3四分位数): データを小さい順に並べたとき、全体の**下から75%**（つまり上から25%）の位置に来る値。
*   `max`: **最大値**。その列で一番大きい値。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/13-15.png)


1.  **列の選択:** 今は全部の数値列の統計量が出てますけど、例えば「15歳未満割合」と「65歳以上割合」の2つだけを比較したい、って時もありますよね。
2.  **数字の見た目:** 小数点以下の桁数がすごく長くて、ちょっと目がチカチカします…。もう少しスッキリさせたいです。
3.  **情報の絞り込み:** 統計量も、平均値と標準偏差、あとは最小・最大値くらいで十分な時もあるかなって。全部表示しなくてもいい場合とか。
4.  **表の向き:** 今は統計量の種類が行で、元のデータの列名が列になってますけど、逆の方が見やすい時もありそうです。変数ごとに行が並んで、その横に平均とか標準偏差が並ぶ、みたいな。
5.  **名前:** やっぱり、行や列の名前が英語のままだと、レポートにするにはちょっと…。日本語にできたら最高です！

**メソッドチェーンとは？**
`pandas` の多くのメソッドは、処理結果として**新しいDataFrameオブジェクトを返す**ように設計されている（非破壊的操作が多かったね）。だから、その返ってきたDataFrameオブジェクトに対して、**さらにドット `.` を付けて次のメソッドを繋げて呼び出す**ことができる。これがメソッドチェーンだ。水が流れるように、処理が次々と繋がっていくイメージだね。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/13-38.png)

```python
# コード例13-19: 【ステップごとの書き方 (再掲)】
# ① 複数の列を抽出
df_selected = df[["15歳未満人口割合", "65歳以上人口割合", "課税対象所得（兆円）"]]

# ② 上の結果 (df_selected) から統計量を計算し、df_desc_for_selected に保存
df_desc_for_selected = df_selected.describe()

# ③ 上の結果 (df_desc_for_selected) を小数点以下2桁で丸め、df_desc_rounded に保存
df_desc_rounded = df_desc_for_selected.round(2)

# 結果を確認
df_desc_rounded
```

```python
# コード例13-20: 【メソッドチェーン (ステップ1と2を連結)】
# ①②　複数の列を抽出し、その結果に直接 .describe() を繋げる！
df_desc_for_selected = df[
    ["15歳未満人口割合", "65歳以上人口割合", "課税対象所得（兆円）"]
].describe()

# ③ 上の結果 (df_desc_for_selected) を小数点以下2桁で丸める
df_desc_rounded = df_desc_for_selected.round(2)

# 結果を確認
df_desc_rounded
```

```python
# コード例13-21: 【メソッドチェーン (ステップ1, 2, 3を連結)】

# 複数の列を抽出し、その結果に .describe() を繋げ、さらにその結果に .round(2) を繋げる！
df_desc_rounded = df[["15歳未満人口割合", "65歳以上人口割合", "課税対象所得（兆円）"]].describe().round(2)

# 結果を確認
df_desc_rounded
```

```python
# コード例13-22
final_stats_chained = df[['15歳未満人口割合', '65歳以上人口割合', '課税対象所得（兆円）']].describe().round(2).loc[['count', 'mean', 'std', 'min', 'max']].transpose().rename(columns={'count': '観測数','mean': '平均値','std': '標準偏差','min': '最小値','max': '最大値'})

# 表示
final_stats_chained
```

### **13.4.3 Excelファイルに注釈を追加する (発展：`openpyxl` の復習)**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/13-49.png)

```python
# コード例13-28

def add_footer(file_path, value, sheet_name=None):
    # file_path (str): Excel ファイルのパス
    # value (str): 注の内容
    # sheet_name (str, optional): シート名。指定しない場合は最初のシートを使用。
    
    import openpyxl

    # Excelファイル（ワークブック）を読み込む
    wb = openpyxl.load_workbook(file_path)

    # シートを取得（指定がなければ最初のシート）
    ws = wb[sheet_name] if sheet_name in wb.sheetnames else wb.worksheets[0]

    # A列の最終行の下のセルに、注の内容を追加
    ws.cell(row=ws.max_row + 1, column=1, value=value)

    # 修正を保存
    wb.save(file_path)
```

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/13-50.png)

```python
# コード例13-29: add_footer 関数にExcelファイルのパスおよび追加したい注釈の内容を渡す
add_footer(
    output_file_path,
    "（注）１）15歳未満人口割合および65歳以上人口割合の単位は%、課税対象所得の単位は兆円。",
)
add_footer(output_file_path, "（出所）筆者作成")
```

### **13.5.1 グループ分けの準備：比較の軸となる「新しい列」を作る**

```python
# コード例13-30: 「高所得地域」列を作るために、課税対象所得（兆円）列を抽出し、中央値を計算
median_income = df["課税対象所得（兆円）"].median()

# 表示
median_income
```

**ステップ3：新しい列を追加し、条件で値を設定する魔法の1行！**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/13-56.png)

**`df_copied['新しい列名'] = 条件比較の結果`**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/13-57.png)

```python
# コード例13-32: '高所得地域' 列を追加
df_copied["高所得地域"] = df_copied["課税対象所得（兆円）"] > median_income

# 最初の5行を表示
df_copied.head()
```

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/13-58.png)

**右辺の解剖：`df_copied['課税対象所得（兆円）'] > median_income`**

*   **`df_copied['課税対象所得（兆円）']`**: まず、`df_copied` というDataFrameから「`課税対象所得（兆円）`」という名前の**列 (Series)** を選び出す。これは、その列の全ての値（各地域の所得）が順番に並んだものだね。
*   **`median_income`**: これはさっき計算した、全地域の「`課税対象所得（兆円）`」の**中央値**（一つの数値）。
*   **`>` (比較演算子)**: ここがポイント！ `pandas` のSeriesは賢くて、この比較演算子 `>` を使うと、Seriesの**各要素**（つまり、各行の「`課税対象所得（兆円）`」の値）と、`median_income` という**一つの数値**を、**一つ一つ順番に比較**してくれるんだ。
*   **比較の結果 (新しいSeries)**: そして、各行ごとの比較の結果（「所得が中央値より大きい」が真か偽か）を、**`True` または `False` の真偽値**として返す。この結果全体も、元のSeriesと同じ行数を持つ、`True` と `False` が並んだ新しい**Series**になるんだ。

### **13.6.2 `regmonkey.summary()` で記述統計量を簡単出力！**

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/13-75.png)

**`summary()` 関数の基本的な使い方:** `summary(データフレーム, var_list=['列名1', '列名2', ...], round_digits=桁数, lang="言語コード")`

*   **最初の引数 (必須):** 分析したい **DataFrame** を指定する。
*   `var_list` 引数 (必須): 統計量を見たい**列名のリスト**を `['列名A', '列名B']` のように指定する。これが以前の `df[['列A', '列B']]` の部分に相当する。
*   `round_digits` 引数 (オプション): 結果の数値を丸める小数点以下の桁数を指定する。デフォルトは `2`（小数点以下2桁）になっている。
*   `lang` 引数 (オプション): 統計量の名前（`count`, `mean` など）をどの言語で表示するか指定する。デフォルトは `'ja'` (日本語) になっている。他にも `'en'` (英語) や `'zh'` (中国語) が選べるぞ。

![](https://raw.githubusercontent.com/qingruikz/python-adventure/main/resources/13-76.png)

```python
# コード例13-42
summary(df, ["15歳未満人口割合", "65歳以上人口割合", "課税対象所得（兆円）"])
```
